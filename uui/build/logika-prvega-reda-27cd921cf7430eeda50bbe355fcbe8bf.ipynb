# Logika prvega reda

V prejšnjem predavanju smo spoznali propozicijsko logiko kot formalni jezik za predstavitev znanja in samodejno sklepanje. Videli smo, da lahko z njo učinkovito opisujemo dejstva o svetu in iz njih izpeljemo sklepe in torej samodejno pridobimo nove informacije. Propozicijska logika nam omogoča uporabo splošnih algoritmov za samodejno sklepanje iz stavkov (disjunkcij literalov), kot je rezolucija, ter učinkovitejše metode, kot sta veriženje naprej in nazaj za poseben primer Hornovih stavkov.

Propozicijska logika ima pomembno omejitev. Njeni osnovni gradniki so atomi, torej osnovne (resnične ali neresnične) izjave brez notranje strukture. Vsako dejstvo moramo zapisati kot samostojen simbol, brez možnosti, da bi v njem vključili bolj splošne objekte, njihove lastnosti ali razmerja med njimi. To postane problematično, ko v bazi znanja želimo vključiti bolj splošno znanje, kot znanje o specifičnih objektih. Omejitve propozicijske logika izhajajo iz omejitev izjavnega računa. V prvem letniku ste spoznali pomembno nadgradnjo izjavnega računa, ki jo imenujemo predikatni račun, ki zato, da lahko predstavimo bolj splošna dejstva, uvaja _predikate_, _spremenljivke_ ter _kvantifikatorja_ $\forall$ in $\exists$.

Logika prvega reda, ki jo bomo obravnavali v tem predavanju, temelji prav na predikatnem računu. Osnovne koncepte iz predikatnega računa bomo zdaj uporabili za formalno definicijo logike prvega reda. Slednjo na področju umetne inteligence uporabljamo kot formalni, nedvoumni jezik za predstavitev splošnega znanja. Spoznali bomo tudi algoritme za samodejno sklepanje iz tako zapisanega znanja.

Razmislimo o naslednjem znanju o študentih in izpitih na nekem namišljenem študijskem programu, ki ima zgolj dva obvezna predmeta. Baza znanja vsebuje pet elementov.

1. Če študent opravi vse obvezne izpite, lahko diplomira.
1. Če študentu mentor potrdi diplomsko nalogo, lahko diplomira.
1. Če študent diplomira, se lahko vpiše na doktorat.
1. Ana je opravila izpit pri predmetu Logika.
1. Ana je opravila izpit pri predmetu Algoritmi.
1. Predmeta Logika in Algoritmi sta obvezna.

Zdaj si lahko zastavimo vprašanje bi samodejno ugotovili pravilen odgovor na vprašanje:

> Ali se Ana lahko vpiše na doktorat?

Nekatere elemente baza znanja iz zgornjega primera lahko predstavimo v propozicijski logiki. Tako, na primer, tretjo in četrto enostavno dejstvo lahko zapišemo z atomoma $\textit{Opravila\_Ana\_Logiko}$ in $\textit{Opravila\_Ana\_Algoritme}$. Težava pri zapisovanju z atomi nastopi pri prvem elementu baze, ki predstavlja bolj splošno pravilo. Za Ano bi lahko to pravilo predstavili kot Hornov stavek
$$ \textit{Opravila\_Ana\_Logiko} \land \textit{Opravila\_Ana\_Algoritme} \implies \textit{Diplomirala\_Ana}.$$
A problem je v tem, da bi v propozicijski logiki morali to pravilo zapisati za vsakega študenta posebej. V relanih primerih bi tako baza znanja dobila na stotine in tisoče Hornovih stavkov z identično strukturo. Poleg tega moramo pravilo prilagoditi tudi vsakemu naboru izpitov za vsak študijski program posebej. Če bi program vseboval več predmetov, bi morali v predpostavke vključiti vse te predmete, kar vodi do zelo dolgih in nepreglednih stavkov in posledično napak pri formalizaciji znanja.

Kot vemo že iz predikatnega računa, v logiki prvega reda lahko isto pravilo zapišemo bistveno bolj elegantno v obliki
$$ \forall x (\textit{Opravil}(x, \textit{Logika}) \land \textit{Opravil}(x, \textit{Algoritmi}) \textit{Potrdil}(\textit{Mentor}(x), x) \implies \textit{Dipomant}(x)),$$
ki velja za vse študente in zgolj za Ano in torej ne zahteva podvojenih zapisov pravil z identično strukturo. Zapis tudi jasno loči med objekti (v tem primeru so objekti študenti, predstavlja jih spremenljivka $x$), funkcijami (funkcija $\textit{Mentor}$ priredi študentki njeno mentorico), njihove lastnosti (ali je študent_ka diplomiral_a, enomestni predikat $\textit{Diplomant}$) ali razmerja med študenti in predmeti (dvomestni predikat \textit{Opravil}). Naj na tem mestu omenimo pomembno razliko med predikati in funkcijami. Predikati opisujejo razmerja med objekti ali lastnosti objektov, funkcije pa danim objektom (ali naborom objektov) priredijo natanko en objekt.

Ostale elemente baze znanja lahko tudi zapišemo z uporabo predikatnega računa:

- $\forall x (\textit{Diplomant}(x) \implies \textit{LahkoVpišeDoktorat}(x))$
- $\textit{Opravil}(\textit{Ana}, \textit{Logika})$
- $\textit{Opravil}(\textit{Ana}, \textit{Algoritmi})$
- $\textit{Obvezen}(\textit{Logika})$
- $\textit{Obvezen}(\textit{Algoritmi})$

Premisli kako bi zgornje pravilo o diplomiranju zapisal krajše: med pogoje ne bi pisal vsak predmet posebej, temveč bolj na kratko zapisal, da je pogoj, da študent_ka opravi *vse* obvezne izpite na njegovem (njenem) študijskem programu. Namig: uporabi predikat $\textit{Obvezen}$ in funkciji $\textit{Program}$, ki študentki (študentu) priredi njen študijski program.

Ključna ideja logike prvega reda je v tem, da znanje zapišemo splošno, ne pa kot množico nepovezanih posameznih primerov. V nadaljevanju predavanja bomo spoznali logiko prvega reda in vsebino osmega poglavja z naslovom _First-Order Logic_ iz učbenika {cite}`russell2019aima`.

## Sintaksa in semantika logike prvega reda

V tem poglavju formalno uvedemo logiko prvega reda kot jezik za predstavitev znanja. Pojmi, ki jih uvedemo, temeljijo na predikatnem računu, ki ste ga spoznali v prvem letniku, vendar jih bomo zapisali v obliki, ki je primerna za samodejno sklepanje.

```{prf:definition} Podpis (angl. _Signature_)
_Podpis_ $\Sigma$ logike prvega reda sestavljajo:

- množica _konstantnih_ simbolov,
- množica _funkcijskih_ simbolov (vsak funkcija, ki ustreza simbolu, ima določeno števnost),
- množica _predikatnih_ simbolov (vsak predikat ima svojo števnost).
```

Konstante predstavljajo posamezne objekte v podpisu (npr. $\textit{Ana}$ ali $\textit{Logika}$), funkcije predstavljajo preslikave med objekti, ki objektu (ali naboru objektov) priredijo točno en objekt (npr. $\textit{Mentor} je enomestna funkcija) in predikati predstavljajo lastnosti objekta ali opisujejo relacije med objekti (npr. $\textit(Obvezni)$ pa enomestni predikat, ki se nanaša na lastnost predmeta, $\textit{Opravil}$ je dvomestni predikat, ki se nanaša na relacijo med študenti in predmeti).

```{prf:definition} Term (angl. _Term_)

_Term_ je izraz, ki predstavlja objekt iz podpisa. Definiran je induktivno:

1. Vsaka _konstanta_ je term,
1. vsaka _spremenljivka_ je term,
1. Če je $f$ funkcijski simbol števnosti $n$ in so $t_1, t_2, \dots, t_n$ termi, je tudi $f(t_1, t_2, \dots, t_n)$ term.
```

Primera termov iz uvodnega razdelka teh zapiskov sta $\textit{x}$, $\textit{Ana}$ in $\textit{Mentor}(x)$. Z drugimi besedami povedano, term je vsak objekt, ki ga predstavimo s konstantnim simbolom ($\textit{Ana}$), spremenljivko ($x$) ali je rezultat funkcije ($\textit{Mentor}(x)$ ali $\textit{Mentor}(Ana)$).

```{prf:definition} Atomska formula (angl. _Atom formula_)

Če je $P$ predikatni simbol števnosti $n$ in so $t_1, t_2, \dots, t_n $ termi, potem je $ P(t_1, t_2, \dots, t_n) $ _atomska formula_.
```

Primera atomskih formul sta $\textit{Opravil}(\textit{Ana}, \textit{Logika})$ ali $\textit{Potrdil}(\textit{Mentor}(x), x)$.

```{prf:definition} Formula logike prvega reda (angl. _First-Order-Logic formula, FOL formula_)

_Fomrulo logike prvega reda_ definiramo induktivno:

1. Vsaka atomska formula je formula,
1. če sta $A$ in $B$ formuli, so formule tudi
    $$
        \neg A,\; A \land B,\; A \lor B,\; A \rightarrow B, 
    $$
1. če je $A$ formula in $x$ spremenljivka, sta formuli tudi
    $$
        \forall x\, A \quad \text{in} \quad \exists x\, A.
    $$
```

Izraz, ki smo ga obravnavali v uvodnem razdelku teh zapiskov,
$$ \forall x (\textit{Opravil}(x, \textit{Logika}) \land \textit{Opravil}(x, \textit{Algoritmi}) \textit{Potrdil}(\textit{Mentor}(x), x) \implies \textit{Dipomant}(x)) $$
je formula v logiki prvega reda.

```{prf:definition} Vezane (angl. _bound_) in proste (angl. _free_) spremenljivke, stavek (angl. _clause_) v logiki prvega reda

Spremenljivka v logični formuli je _vezana_, če nastopa pod kvantifikatorjem, sicer je _prosta_.

_Stavek_ (angl. _clause_) je logična formula v logiki prveg reda, ki nima prostih spremenljivk.
```

Premisli ali je spremenljivka $x$ v izrazu navedenem pred to definicijo vezana ali prosta. Pojasni zakaj. Katera od logičnih formul $\textit{Opravil}(x, \textit{Logika})$ in $\forall x (\textit{Opravil}(x, \textit{Logika}))$ je stavek v logiki prvega reda in zakaj?

Zdaj, ko smo definirali sintakso logičnih formul (in stavkov) v logiki prvega reda, lahko se lotimo definicije njihovega pomena. Za razliko od propozicijske logike, kjer interpretiramo le resničnost atomov in njihovih kombinacij, ki nastajajo z uporabo logičnih veznikov (logičnih operacij), moramo v logiki prvega reda **interpretirati** tudi strukturo formul.

```{prf:definition} Interpretacija (angl. _Interpretation_)

_Interpretacijo_ $\mathcal{I}$ podaja:

- neprazna množica $D$, ki jo imenujemo domena,
- preslikava, ki:
  - vsakemu konstantnemu simbolu priredi element iz $D$,
  - vsakemu funkcijskemu simbolu števnosti $n$ priredi funkcijo $D^n \to D$,
  - vsakemu predikatnemu simbolu števnosti $n$ priredi relacijo na $D$, ki je podmnožica kartezičnega produkta $\underbrace{D \times D \times \dots \times D}_{n \, \text{krat}}$
```

Možna domena za primer iz uvodnega razdelka teh zapiskov je $D = \{ \textit{Ana}, \textit{Logika}, \textit{Algoritmi} \}$. Kot vidimo iz te domene, Logika prvega reda ne ločuje tipov objektov. Vsi objekti so v isti domeni, njihov tip, ko ga potrebujemo, lahko določimo kot lastnost objekta, torej z ustreznim enometnim predikatom. V tem primeru bi potrebovali dva taka predikata $\textit{Student}$ in $\textit{Predmet}$ in tri atomske formule:
- $\textit{Student}(\textit{Ana})$,
- $\textit{Predmet}(\textit{Logika})$ in
- $\textit{Predmet}(\textit{Algoritmi})$.

Ker formule v logiki prvega reda vsebujejo spremenljivke, moramo pri njihovem pomenu uporabljati še vrednotenje, ki ga definiramo takole.

```{prf:definition} Vrednotenje

_Vrednotenje_ je preslikava, ki vsaki spremenljivki priredi element iz domene intepretacije.
```

Vzamimo za formulo $\forall x (\textit{Opravil}(x, y))$. Vrednotenje za to formulo je preslikava $v \colon \{ x, y \} \to D$, kjer je $D$ domena podane intepretacije $\mathcal{I}$. Primer vrednotenja je preslikava definirana s predpisi $v \colon x \mapsto \textit{Ana}$ (zapišemo tudi $v(x) = \textit{Ana}$) in $v \colon y \mapsto \textit{Logika}$ (zapišemo tudi $v(y) = \textit{Logika}$). Vrednotenje torej določa, kateri objekti iz domene $D$ ustrezajo spremenljivkam podane logične formule. Notacija $v[x \mapsto d]$ pomeni, da vrednost spremenljivke $x$ zamenjamo z elementom $d$ iz domene $D$, $d \in D$ interpretacije $\mathcal{I}$.

Premisli kako vrednotenje vpliva na vezane spremenljivke formule v logiki prvega reda. V primeru iz prejšnjega odstavka premisli kakšen je vpliv predpisa $v(x) = \textit{Ana}$ na intepretacijo formule $\forall x (\textit{Opravil}(x, y))$.

S pomočjo interpretacije in vrednotenja lahko končno definiramo resničnost formule v logiki prvega reda. Pred tem seveda, zaradi prisotnosti spremenljivk v formuli prvega reda, moramo previdno in postopoma defincirati veljavnost v podani interpretaciji (in z njo povezani domeni) in podanem vrednotenju (prirejanju spremenljivk).

```{prf:definition} Zadoščanje

Naj bo $\mathcal{I}$ interpretacija z domeno $D$ in $v$ vrednotenje spremenljivk. Pravimo, da _interpretacija $\mathcal{I}$ pri vrednotenju $v$ *zadosti* formuli $A$_, in pišemo
$$ \mathcal{I}, v \models A, $$
če velja:

- $\mathcal{I}, v \models P(t_1,\dots,t_n)$ natanko tedaj, ko $ (t_1^{\mathcal{I}, v}, t_2^{\mathcal{I}, v}, \dots, t_n^{\mathcal{I},v}) \in P^\mathcal{I}$,
- $\mathcal{I}, v \models \neg A$ natanko tedaj, ko $\mathcal{I}, v \not\models A$,
- $\mathcal{I}, v \models A \land B$ natanko tedaj, ko $\mathcal{I}, v \models A$ **in** $\mathcal{I}, v \models B$,
- $\mathcal{I}, v \models A \lor B$ natanko tedaj, ko $\mathcal{I}, v \models A$ **ali** $\mathcal{I}, v \models B$,
- $\mathcal{I}, v \models A \implies B$ natanko tedaj, ko $\mathcal{I}, v \not\models A$ ali $\mathcal{I}, v \models B$,
- $\mathcal{I}, v \models \forall x\, A$ natanko tedaj, ko za vsak $d \in D$ velja $\mathcal{I}, v[x \mapsto d] \models A$,
- $\mathcal{I}, v \models \exists x\, A$ natanko tedaj, ko obstaja $d \in D$ za katerega velja $\mathcal{I}, v[x \mapsto d] \models A$.
```

PRIMER, nadaljevanje. Poglejmo pravilo za univerzalni kvantifikator in ugotovimo, da vezana spremenljivka ne potrebuje vrednotenja. Podobno velja tudi za eksistenčni kvantifikator.

Primer nam nazorno pokaže, da pri preverjanju zadoščanja, so iz vrednotenja pomembne samo vrednosti prostih spremenljivk. V stavkih, ki so logične formule brez prostih spremenljivk, vrednotenje ne igra vloge. V takih primerih zadoščanje imenujemo model, kot kaže naslednja definicija.

```{prf:definition} Model

_Interpretacija $\mathcal{I}$ je model formule $A$_, če velja
$$
\mathcal{I} \models A,
$$
kar pomeni, da $\mathcal{I}, v \models A$ za vsako možno vrednotenje $v$.
```

```{prf:definition} Veljavnost

Formula $A$ v logiki prvega reda je _veljavna_ natanko tedaj, ko je vsaka interpretacija njen model.
```

Na tem mestu bi bilo pomembno poudariti razlike med zadnjimi definicijami.
- $\mathcal{I}, v \models A$ beremo _interpretacija $\mathcal{I}$ zadosti (angl. *satisfies*) formuli $A$ pri vrednotenju $v$.
- $\mathcal{I} \models A$ beremo _interpretacija $\mathcal{I}$ je model (angl. *is a model of*) formule $A$.
- Formula $A$ je _veljavna_ (angl. _valid_), če je vsaka interpretacija $\mathcal{I}$ model formule $A$, oziroma vsaka interpretacija $\mathcal{I}$ zadosti formuli $A$ pri vsakem vrednotenju $v$.
- Stavek $A$ je _veljaven_, če je vsaka interpretacija $\mathcal{I}$ model stavka $A$. Ne pozabimo, da stavek nima prostih spremenljivk in zato izbira vrednotenja ne vpliva na njegovo veljavnost.

Zdaj, ko vemo, kaj je veljavna formula lahko končno definiramo sklepanje v logiki prvega reda.

```{prf:definition} Sklepanje v logiki prvega reda

Naj bodo $A_1, A_2, \dots, A_n$ in $B$ formule v logiki prvega reda in naj bo $\mathcal{A} = \{ A_1, A_2, \dots, A_n \}$. Relaciji
$$
\mathcal{A} \models B
$$

rečemo _sklepanje_. Elemente množice $\mathcal{A}$ imenujemo _predpostavke_, formulo $B$ pa _sklep_ ali _zaključek sklepanja_.
```

Zdaj povežemo veljavnost sklepanja z veljavnostjo formul v logiki prvega reda.

```{prf:definition} Veljavnost sklepanja v logiki prvega reda

Sklepanje $\mathcal{A} \models B$ je _veljavno_, če za vsako interpretacijo $\mathcal{I}$ velja:

če $\mathcal{I} \models A$ za vse $A \in \mathcal{A}$, potem velja tudi $\mathcal{I} \models B$.
```

Končno lahko definiramo logično posledico v logiki prvega reda takole.

```{pfr:definition} Logična posledica v logiki prvega reda

Če je sklepanje $\mathcal{A} \models B$ veljavno, potem rečemo, da je $B$ _logična posledica_ množice predpostavk $\mathcal{A}$.
```

Spomnimo se prejšnjega predavanja. V propozicijski logiki smo veljavnost sklepanja preverjali tako, da smo preverili, če je formula $(A_1 \land A_2 \land \dots A_n) \implies B$ tavtologija. Pomembna razlika pri sklepanju v logiki prvega reda, da njegovo veljavnost definiramo s pomočjo interpretacij. Sklepanje je veljavno, če v vsaki interpretaciji $\mathcal{I}$, ki zadosti predpostavkam iz $\mathcal{A}$, zadosti tudi sklepu $B$. Namreč, v logiki prvega reda formule niso več tavtologije v enakem smislu kot v propozicijski logiki, zato veljavnost sklepanja definiramo preko modelov in interpretacij.

Da bi lahko sklepanje v logiki prvega reda izvedli z rezolucijo, moramo najprej formule prevesti v ustrezno obliko in obvladati ujemanje izrazov. V naslednjih dveh poglavjih bomo zato najprej spoznali skolemizacijo in unifikacijo, ki predstavljata ključni komponenti rezolucijskega dokazovanja v logiki prvega reda

## Pretvorba formul logike prvega reda v stavčno konjunktivno normalno obliko in skolemizacija

Tako kot pri sklepanju v propozicijski logiki najprej moramo poskrbeti za pretvorbo formul v konjunktivno normalno obliko. Zaradi kvantifikatorjev in spremenljivk je postopek tokrat bolj zapleten kot postopek ustrezne pretvorbe izjavne formule v propozicijski logiki. V logiki prvega reda sledimo naslednjim korakom:

1. odstranjevanje implikacij,
1. premik negacij,
1. standardizacija spremenljivk,
1. tvorjenje preneksne oblike,
1. skolemizacija in
1. odstranitev univerzalnih kvantifikatorjev.

Korake bomo vpeljevali enega za drugim in vsak korak bomo ilustrirali na primeru pretvorbe formule v logiki prvega reda
$$
    \forall x\, \left( \textit{Student}(x) \implies \exists y \, \left( \textit{Opravil}(x,y) \land \neg \forall x \, \textit{Obrajta}(x, y) \right) \right)
$$
v njeno stavčno konjunktivno normalno obliko.

Tako kot v propozicijski logiki pri _odstranjevanju implikacij_ sledimo preprosti pretvorbi implikacije $A \implies B$ v disjunkcijo $\neg A \lor B$. V zgornji formuli se implikacija pojavi enkrat in jo nadomestimo takole:
$$
    \forall x\, \left( \neg \textit{Student}(x) \lor \exists y \, \left( \textit{Opravil}(x,y) \land \neg \forall x \, \textit{Obrajta}(x, y) \right) \right).
$$

V drugem koraku s _premikom negacij_ poskrbimo, da se negacije pojavljajo neposredno pred atomskimi formulami. Negacije, ki se pojavijo kjerkoli v formuli, sistematično premikamo navznoter, dokler ne stojijo neposredno pred posameznimi predikati. Pri tem uporabljamo De Morganova zakona za konjunkcijo in disjunkcijo ter pravila za premik negacije čez kvantifikatorje. Spomnimo se najprej De Morganovih zakonov.

1. De Morganov zakon za konjunkcijo pove, da je formula $\neg (A \land B)$ ekvivalentna formuli $\neg A \lor \neg B$.
1. De Morganov zakon za disjunkcijo pove, da je formula $\neg (A \lor B)$ ekvivalentna formuli $\neg A \land \neg B$.

Negacijo lahko premikamo tudi čez kvantifikatorje. Pri tem velja pravilo, ki ga lahko razumemo kot posplošitev De Morganovega zakona, ki pove, da je formula $\neg \forall x\, A$ ekvivalentna formuli $\exists x\, \neg A$. Ta ekvivalenca izraža naslednjo intuitivno idejo: izjava "formula $A$ ne velja za vsako vrednost spremenljivke $x$" je ekvivalentna izjavi "obstaja vrednost spremenljivke $x$ za katero velja formula $\neg A$ (oziroma ne velja formula $A$)".

Podobno lahko posplošimo tudi drugi zakon za disjunkcijo: formula $\neg \exists x\, A$ je ekvivalentna formuli $\forall x\, \neg A$. Z besedami povedano izjava "ne obstaja vrednost spremenljivke $x$ za katero velja formula $A$" je ekvivalentna izjavi "formula $\neg A$ velja za vsako vrednost spremenljivke $x$". V zgornji formuli moramo premakniti negacijo pred zadnjim univerzalnim kvantifikatorjem takole:
$$
    \forall x\, \left( \neg \textit{Student}(x) \lor \exists y \, \left( \textit{Opravil}(x,y) \land \exists x \, \neg \textit{Obrajta}(x, y) \right) \right).
$$

V tretjem koraku poskrbimo za _standardizacijo spremenljivk_, tako da preimenujemo vezane spremenljivke v formuli. Če ista spremenljivka nastopa enkrat kot prosta in drugič kot vezana, ali če je ista spremenljivka vezana z različnimi kvantifikatorji, moramo njene vezane pojavitve preimenovati. Običajno vezano spremenljivko nadomestimo z novo spremenljivko, ki še ne nastopa nikjer drugje v formuli. Standardizacija zagotovi, da vsak kvantifikator "upravlja" svojo spremenljivko. V našem primeru formule se spremenljivka $x$ pojavi dvakrat. Temu se izognemo tako, da drugo pojavitev preimenujemo v $z$:
$$
    \forall x\, \left( \neg \textit{Student}(x) \lor \exists y \, \left( \textit{Opravil}(x,y) \land \exists z \, \neg \textit{Obrajta}(z, y) \right) \right).
$$

V četrtem koraku poskrbimo za pretvorbo formule v _preneksno obliko_. Za formulo logike prvega reda rečemo, da je v _preneksni obliki_, če ima obliko $Q_1 x_1 \; Q_2 x_2 \; \dots \; Q_n x_n \; A$, kjer so $Q_i \in \{\forall, \exists\}$ kvantifikatorji, $x_i$ spremenljivke, $A$ pa formula brez kvantifikatorjev. V taki formuli so vsi kvantifikatorji na začetku formule, preostanek formule pa je "čisti" logičen izraz brez kvantifikatorjev. Pri pretvorbi v preneksno obliko torej gre za premik kvantifikatorjev na začetek formule, kar nam omogočajo naslednja štiri pravila.

1. Formula $(\forall x\, A) \lor B$ je ekvivalentna formuli $\forall x\, (A \lor B)$.
1. Formula $(\exists x\, A) \lor B$ je ekvivalentna formuli $\exists x\, (A \lor B)$.
1. Formula $(\forall x\, A) \land B$ je ekvivalentna formuli $\forall x\, (A \land B)$.
1. Formula $(\exists x\, A) \land B$ je ekvivalentna formuli $\exists x\, (A \land B)$.

Ker smo v prejšnjem koraku izvedli standardizacijo spremenljivk, lahko kvantifikatorje premikamo čez logične veznike brez dodatnih pogojev, ne da bi pri tem spremenili pomen formule. Sicer pa za tovrstne premike velja pogoj, da spremenljivka $x$ ne nastopa prosto v $B$. Ta pogoj zagotavlja, da pri premiku kvantifikatorja ne spremenimo pomena formule.

Z nekaj takih premikov se naša formula spremeni v
$$
    \forall x\, \exists y\, \exists z\, \left( \neg \textit{Student}(x) \lor \left( \textit{Opravil}(x,y) \land \neg \textit{Obrajta}(z, y) \right) \right).
$$

V petem koraku izvedemo _skolemizacijo_, s katero odstranimo eksistenčne kvantifikatorje iz formule. Osnovna ideja skolemizacije je, da eksistenčne spremenljivke nadomestimo z novimi simboli, ki predstavljajo njihove vrednosti.

Če se eksistenčni kvantifikator pojavi izven dosega univerzalnih kvantifikatorjev, pripadajočo spremenljivko nadomestimo z novo konstanto. Če pa se eksistenčni kvantifikator pojavi v dosegu univerzalnih kvantifikatorjev, spremenljivko nadomestimo s funkcijo teh univerzalnih spremenljivk.

Natančneje, naj bo formula v preneksni obliki. Vsako pojavitev podformule oblike $\exists x\, A$ obravnavamo na naslednji način:

- če $x$ ni v dosegu nobenega univerzalnega kvantifikatorja, ga nadomestimo z novo konstanto $c$
- če je $x$ v dosegu univerzalnih kvantifikatorjev nad spremenljivkami $u_1, u_2, \dots, u_n$, ga nadomestimo z novim funkcijskim simbolom $f(u_1, u_2, \dots, u_n)$.

Po tej zamenjavi eksistenčni kvantifikator odstranimo. S tovrstnimi zamenjavami dobimo formulo, ki ne vsebuje eksistenčnih kvantifikatorjev. Pomembno je poudariti, da skolemizacija ne ohranja logične ekvivalentnosti, temveč ohranja _neskladnost_ formule. Intuitivno namreč skolemizacija zgolj nadomesti izjavo oblike "obstaja nek element" z izjavo "ta element je podan s funkcijo (ali konstanto)". Če nobena izbira ne zadošča logični formuli, potem tudi tako določena izbira ne bo, zato skolemizacija ohranja neskladnost.

Pri skolemizaciji naše formule moramo nadomestiti dve eksistenčni spremenljivki $y$ in $z$. Kar sta obe v dosegu univerzalnega kvantifikatorja $\forall x$ ju zamenjamo s funkcijskimi simboli $f(x)$ in $g(x)$:
$$
    \forall x\, \left( \neg \textit{Student}(x) \lor \left( \textit{Opravil}(x, f(x)) \land \neg \textit{Obrajta}(g(x), f(x)) \right) \right).
$$

Po skolemizaciji dobimo formulo, ki ne vsebuje eksistenčnih kvantifikatorjev in je v preneksni obliki. V zadnjem, šestem koraku odstranimo še univerzalne kvantifikatorje. Ker so v stavčni obliki vse spremenljivke implicitno univerzalno kvantificirane, lahko kvantifikatorje $\forall$ preprosto izpustimo. Na našem primeru s tem dobimo stavčno obliko:
$$
    \neg \textit{Student}(x) \lor \left( \textit{Opravil}(x, f(x)) \land \neg \textit{Obrajta}(g(x), f(x)) \right).
$$

Tako dobimo formulo, ki vsebuje le še logične veznike in spremenljivke. Preostane nam, da formulo pretvorimo v konjunktivno normalno obliko, torej v konjunkcijo disjunkcij literalov. Pri tem uporabimo enake postopke kot v propozicijski logiki. V našem primeru potrebujemo distribucijski zakon, ki pove, da je formula $A \lor (B \land C)$ ekvivalentna formuli $(A \lor B) \land (A \lor C)$:
$$
    \left( \neg \textit{Student}(x) \lor \textit{Opravil}(x, f(x)) \right)
    \land
    \left( \neg \textit{Student}(x) \lor \neg \textit{Obrajta}(g(x), f(x)) \right).
$$

Končno formulo v konjunktivni normalni obliki, tako kot v primeru propozicijske logike, interpretiramo kot množico stavkov, kjer vsaka disjunkcija literalov predstavlja en stavek. Ti stavki predstavljajo vhod za rezolucijsko dokazovanje v logiki prvega reda. Na koncu torej za naš primer dobimo množici $\{ \neg \textit{Student}(x), \textit{Opravil}(x, f(x)) \}$ in $\{ \neg \textit{Student}(x), \neg \textit{Obrajta}(g(x), f(x)) \}$. Od te točke naprej lahko nad klavzulami uporabljamo rezolucijo — edina nova stvar, ki jo še potrebujemo, je unifikacija.

## Unifikacija

V pripravi.

```{bibliography}
```
