# Propozicijska logika

Do sedaj smo na predavanjih obravnavali različne pristope k reševanju problemov v umetni inteligenci, ki temeljijo na preiskovanju. Najprej smo spoznali splošne preiskovalne algoritme, pri katerih problem predstavimo kot prostor stanj, algoritem pa sistematično preiskuje možne poti od začetnega do enega izmed ciljnih stanj. Ugotovili smo, da so preiskovalni algoritmi uporabni tudi pri izbiri optimalnih potez v igrah. Pri problemih, kjer strukturo problema opišemo bolj neposredno z omejitvami med spremenljivkami, uporabimo algoritme za zadoščanje omejitvam. Cilj preiskovanja v teh primerih je dodelitev vrednosti spremenljivkam, ki zadoščajo podanim omejitvam. 

Pristopi si delijo pomembno lastnost. Problem rešujemo predvsem tako, da preiskujemo prostor možnih rešitev, pri čemer nam struktura problema pomaga zmanjšati število možnosti, ki jih moramo preveriti. Ljudje pri reševanju problemov pogosto imamo na voljo še nekaj več. To je splošno znanje o svetu, zapisano v obliki pravil ali zakonitosti. V takih primerih lahko nove informacije dobimo tako, da iz znanih dejstev logično _sklepamo_, namesto da bi zgolj preiskovali prostor možnih rešitev in stanj. Algoritmi, ki takšno sklepanje izvajajo samodejno, spadajo na področje samodejnega sklepanja (angl. _autoamted inference_).

Matematična logika je priročen formalni jezik za predstavitev znanja in okvir za sklepanje. V tem predavanju bomo spoznali _izjavno logiko_, ki jo bomo, zaradi usklajenosti z literaturo s področja umetne inteligence, _propozicijska logika_ (od angl. _propositional logic_). _Izjavni račun_, ki ste ga spoznali v prvem letniku, temelji na izjavni logiki. Tam ste tudi spoznali osnove pristope k samodejnemu dokazovanju, ki je temelj za pravilno samodejno sklepanje. Namreč, sklepanje je izpeljava novih dejstev iz že znanih, dokazovanje pa je formalni postopek za utemeljitev sklepanja. V okviru tega predavanja bomo spoznali učinkovite algoritme za samodejno sklepanje in omejen nabor logičnih stavkov, ki zagotavljajo večjo učinkovitost teh algoritmov.

Poglejmo zdaj primer uganke, ki jo bomo reševali s samodejnim sklepanjem. V raziskovalnem laboratoriju proizvodnega podjetja je izginil prototip. Osumljeni so Anže, Brina in Ciril. Detektiv po pogovoru z zaposlenimi v podjetju ugotovi naslednji nabor dejstev.

  1. Prototip je ukradel vsaj eden od treh osumljencev.
  1. Če je prototip ukradel Anže, potem bi bili izbrisani dnevniški zapisi.
  1. Če je prototip ukradla Brina, potem bi bilo razbito okno.
  1. Dnevniški zapisi niso bili izbrisani.
  1. Okno ni bilo razbito.

Vprašanje, ki zahteva sklepanje iz podanih dejstev:

> Kdo je ukradel prototip?

Premisli kdo izmed treh osumljencev je ukradel prototip. Utemelji svoj odgovor. Na osnovi znanja o izjavnem računu iz prvega letnika ugotovi, kako bi to uganko predstavil v izjavni (propozicijski) logiki. Kako bi uporabil izjavni račun za dokazovanje krivde ali nedolžnosti treh osumljencev? Kako se spremeni sklep o kraji prototipa, če iz ugotovljenih dejstev odstranimo zadnje (o tem, da okno ni bilo razbito)?

Za reševanje te uganke s samodejnim sklepanjem, bomo v okviru predavanja spoznali vsebino sedmega poglavja z naslovom _Propositional Theorem Proving_ iz učbenika {cite}`russell2019aima`. Omejili se bomo na vsebino razdelkov tega poglavja do (vključno) 7.5.

## Izjavni račun, sklepanje in avtomatsko dokazovanje

Ponovimo nekaj definicij osnovnih konceptov, ki jih že poznate iz matematične logike in izjavnega računa. Bodite pozorni na poimenovanje teh konceptov, ki je lahko v kontekstu umetne inteligence drugačno, čeprav gre za iste koncepte.

```{prf:definition} Atom, literal in stavek
:label: def-osnovne

_Atom_ (ali atomska logična formula) je elementarno dejstvo v propozicijski logiki, ki ne vsebuje logičnih veznikov in ima lahko le eno izmed dveh logičnih vrednosti, resnično (oznaki `True` ali $\top$) ali neresnično (oznaki `False` ali $\bot$). Za oznake atomov bomo uporabljali velike latinske črke.

_Literal_ je poljubni atom $A$ ali njegova negacija $\neg A$.

_Stavek_ je disjunkcija literalov $L_1 \lor L_2 \lor \dots \lor L_n$. Lahko ga zapišemo tudi v obliki množice literalov v disjunkciji, $\{ L_1, L_2, \dots, L_n \}$.
```

Pozor, v matematični logiki, atomu običajno rečemo _izjava_. Če ima atom resnično vrednost, rečemo tudi, da je izjava pravilna, sicer pa rečemo, da je izjava nepravilna.

_Logični vezniki_ ustrezajo logičnim operacijam nad atomi. Vključujejo unarno operacijo negacije $\neg$, ter binarne operacije konjunkcije $\land$, disjunkcije $\lor$, implikacije $\implies$ in ekvivalence $\iff$. Pomen (semantiko) teh logičnih operacij v izjavnem računu definiramo z resničnostnimi tabelami. V primeru negacije se tabela nanaša na en atom in ima zato dve vrstici (eno, ki ustreza vrednosti atoma $\top$ in eno za vrednost atoma $\bot$). V primeru binarnih operacij imajo ustrezne tabele štiri vrstice, vsaka od njih se nanaša na kombinacijo logičnih vrednosti dveh atomov.

Iz zapiskov predavanj v prvem letniku ugotovi kakšne so resničnostne tabele za navedene logične operacije. Premisli koliko je _pomensko različnih_ unarnih in binarnih logičnih operacij?

```{prf:definition} Logična formula

  1. Vsaka logična vrednost $\top$ ali $\bot$ je logična formula.
  1. Vsak literal je logična formula.
  1. Če sta $A$ in $B$ stavka in je $\circ$ poljuben birarni logični veznik, je tudi $A \circ B$ logična formula.

_Spremenljivke logične formule_ so elementi množice atomov, ki nastopajo v logični formuli. Logična formula je _tavtologija_, če ima resnično vrednost za vsak možen nabor vrednosti svojih spremenljivk. Logična formula je _neskladna_, če ima neresnično vrednost za vsak možen nabor vrednosti svojih spremenljivk.
```

Pozor, v matematični logiki logični formuli rečemo tudi _sestavljena izjava_ ali _izjavni izraz_.

Iz predavanj v prvem letniku vemo, da lahko vsaki logični formuli priredimo resničnostno tabelo. Vsaka vrstica v resničnostni tabeli ustreza _določilu_, ki je nabor vrednosti spremenljivk logične formule. Premisli koliko vrstic (določil) bo imela resničnostna tabela za stavek $P \implies Q \land R$, kjer so $P$, $Q$ in $R$ atomi. Zapiši resničnostno tabelo za ta izraz. Iz zapiskov za predavanja v prvem letniku ugotovi kakšna je prioriteta logičnih operacij, torej ugotovi ali je pravilen pomen podanega stavka $(P \implies Q)\land R$ ali $P \implies (Q \land R)$.

Logična formula $P$ je lahko podana tudi z resničnostno tabelo. Iz podane resničnostne tabele lahko izpeljemo logično formulo tako, da iz tabele sestavimo njeno _konjunktivno normalno obliko_ (v izjavnem računu temu rečemo tudi _izbrana konjunktivna oblika_, v angleščini _conjunctive normal form, CNF_). Spomnimo se, da vsakemu določilu iz resničnostne tabele, kjer je vrednost formule $P$ neresnična ($\bot$), priredimo stavek $D$. Vsak literal v stavku $D$ ustreza eni izmed spremenljivk $A$ logične formule $P$. Če je vrednost spremenljivke $A$ v določilu $\top$, potem v stavek $D$ dodamo literal $\neg A$, sicer pa stavku $D$ dodamo literal $A$. Logični formuli, ki je konjunkcija vseh stavkov, ki smo jih na ta način sestavili iz podane resničnostne tabele za $P$, rečemo _konjunktivna normalna oblika_ logične formule $P$.

Za nadaljevanje predavanj je bistvenega pomena, da si zapomnimo dvoje. Najprej, konjunkitvna normalna forma logične formule je konjunkcija stavkov. Stavek je logična formula, ki jo lahko zapišemo v konjunktivno normalno formo z enim samim konjunktom.

Premisli ali iz zapiskov za predavanja v prvem letniku ugotovi katerim logičnim formulam ne moremo prirediti konjunktivne normalne oblike. Ugotovi tudi kako iz resničnostne tabele sestavimo _disjunktivno normalno obliko_ (ali _izbrano disjunktivno obliko_) logične formule in katerim formulam ne moremo prirediti te oblike.

```{prf:definition} Sklepanje

Naj bodo $A_1, A_2, \dots, A_n$ in $B$ logične formule in naj bo $\mathcal{A} = \{ A_1, A_2, \dots, A_n \}$. Logični formuli $S = (A_1 \land A_2 \land \dots \land A_n) \implies B$, ki jo krajše zapišemo v obliko $\mathcal{A} \models B$, rečemo _sklepanje_. Elemente množice $\mathcal{A}$ imenujemo _predpostavke_, $B$ pa _sklep_ ali _zaključek sklepanja_.

Če je logična formula $S$ tavtologija, potem rečemo, da je _sklepanje veljavno_, sicer je pa _sklepanje neveljavno_. Če je sklepanje veljavno, imenujemo sklep $B$ _logična posledica_ množice predpostavk $\mathcal{A}$.
```

Glede na definicijo sklepanja, lahko s hitrim premislekom o tem, kdaj je leva stran implikacije sklepanja resnična in na osnovi pomena (resničnostne tabele) operacije implikacija, dokažemo naslednji izrek.

```{prf:theorem} Logične vrednsti predpostavk in sklepa pri veljavnem sklepanju

Naj bo $\mathcal{A} \models B$ veljavno sklepanje. Če pri nekem določilu za logično formulo sklepanja (naboru logičnih vrednosti spremenljivk logične formule sklepanja) vse predpostavke iz $\mathcal{A}$ dobijo vrednost $\top$, potem pri istem določilu tudi sklep $B$ dobi logično vrednost $\top$.
```

Izrek pojasni pomen veljavnega sklepanja: sklepanje in sklep sta veljavna, če je sklep resničen takrat, ko so resnične predpostavke. Zdaj lahko povežemo pojem veljavnega sklepanja s problemom samodejnega (avtomatskega) dokazovanja.

```{prf:definition} Samodejno dokazovanje

_Samodejno dokazovanje_ je problem samodejnega, torej avtomatskega ugotavljanja, ali je podani zaključek $B$ logična posledica podane množice predpostavk $\mathcal{A}$, oziroma ali je sklepanje $\mathcal{A} \models B$ veljavno. Pri samodejnem dokazovanju množico predpostavk $\mathcal{A}$ imenujemo tudi _baza znanja_ (angl. _knowledge base, KB_).
```

Samodejno dokazovanje lahko izpeljemo tako, da opazujemo resničnostno tabelo logične formule sklepanja $(A_1 \land A_2 \land \dots \land A_n) \implies B$ in ugotavljamo, če je formula tavtologija oziroma velja, da vsakič, ko so resnične predpostavke, je resničen tudi sklep $B$ (glej izrek zgoraj). Tak način dokazovanja je nepraktičen, ker število vrstic resničnostne tabele eksponentno narašča s številom spremenljivk logične formule sklepanja.

Drug način samodejnega dokazovanja, ki ga poznamo iz izjavnega računa, je logična izpeljava sklepa $B$ iz predpostavk. Pri logični izpeljavi uporabimo zaporedje izpeljevalnih pravil (veljavnih sklepanj), ki iz podanih predpostavk izpeljemo sklep. Primer izpeljevalnega pravila, ki ga imenujemo _modus ponens_ je
$$
\dfrac{A \qquad A \implies B}{B}
\quad\text{(modus ponens)}.
$$
Pravilo določi, da iz dveh logičnih formul $A$ in $A \implies B$ lahko izpeljemo $B$. Nabor izpeljevalnih pravil lahko najdemo v zapiskih za predavanja iz izjavnega računa.

S pomočjo izpeljevalnih pravil lahko definiramo preiskovalni prostor, kjer stanja ustrezajo stavkom, začetno stanje vsebuje konjunkcijo predpostavk, končno stanje ustreza sklepu, funkcijo prehodov pa definiramo na osnovi izpeljevalnih pravil. Rešitev tako definiranega preiskovalnega problema bi bila logična izpeljava sklepa, če ta obstaja. Ta način samodejnega dokazovanja je tudi nepraktičen, saj ima tako definiran prostor stanj visoko stopnjo razvejanosti (premisli čemu je enaka stopnja razvejanosti tega prostora) in je težko definirati hevristiko, ki bi povečala učinkovitost preiskovanja.

Tretji pristop k reševanju samodejnega dokazovanja je dokazovanje s protislovjem: namesto da bi neposredno izpeljevali zaključek, predpostavimo njegovo negacijo in preverimo, ali skupaj s predpostavkami vodi v protislovje. Na tej ideji temelji _rezolucija_, ki je, zaradi svoje enostavnosti in učinkovitosti, osrednja metoda samodejnega dokazovanja v propozicijski logiki.

## Dokazovanje z rezolucijo

Formalizirajmo najprej dokazovanje s protislovjem pri sklepanju. Spomnimo se, da pri preverjanju veljavnosti sklepanja $\mathcal{A} = \{ A_1, A_2, \dots, A_n \} \models B$ obravnavamo logično formulo $(A_1 \land A_2 \land \dots \land A_n) \implies B$. Pri dokazovanju te formule s protislovjem, predpostavimo, da $B$ ni resničen, kar je ekvivalentno predpostavki, da je literal $\neg B$ resničen. Torej je logična formula, ki jo obravnavamo na začetku dokazovanja veljavnosti sklepanja s protislovjem, konjunkcija hipotez in negacije sklepa, $A_1 \land A_2 \land \dots \land A_n \land \neg B$. Sklepanje je veljavno natanko tedaj, ko je ta stavek neskladen (vedno neresničen). Torej, če iz množice $\{ A_1, A_2, \dots, A_n, \neg B \}$ lahko izpeljemo protislovje, potem je sklepanje $\mathcal{A} \models B$ veljavno.

Pri logični izpeljavi protislovja, uporabljamo naslednje rezolucijsko pravilo, ki nam omogoča, da iz dveh stavkov, ki vsebujeta komplementarna literala 
$L$ in $\neg L$, izpeljemo nov stavek, v katerem ta literala izločimo, kot kaže naslednja definicija.

```{prf:definition} Rezolucijsko pravilo

$$
\dfrac{L \lor A \qquad \neg L \lor B}{A \lor B}
\quad\text{(rezolucija)}
$$
Naj bosta $S_1 = L \lor A$ in $S_2 = \neg L \lor B$ stavka, kjer je $L$ literal, $A$ in $B$ sta pa (lahko tudi prazni) disjunkciji literalov. _Rezolventa_ stavkov $S_1$ in $S_2$ glede na literal $L$ je stavek $A \lor B$. Pravilu, ki iz $S_1$ in $S_2$ izpelje nujino rezolvento pravimo _rezolucijsko pravilo_.

Če sta stavka $S_1$ in $S_2$ obravnavamo kot množici literalov, potem rezolventa ustreza množici literalov $\left(S_1 \setminus \{ L \}\right) \cup \left(S_2 \setminus \{ \neg L \}\right)$.
```

Premislimo zdaj, kaj se zgodi, če z rezolucijskim pravilom izpeljemo prazen stavek (ki ustreza prazni množici literalov). Vsak stavek $L_1 \lor L_2 \lor \dots \lor L_n$ lahko zapišemo kot $\bot \lor L_1 \lor L_2 \lor \dots \lor L_n$. Torej je prazen stavek, kjer je $n = 0$, pravzaprav $\bot$. Kar je enako protislovju. Zato naslednja definicija.

```{prf:definition} Rezolucijski dokaz

Rezolucijski dokaz je zaporedje stavkov, kjer je vsak stavek bodisi eden izmed začetnih stavkov bodisi rezolventa dveh predhodnih stavkov. Če v takem zaporedju dobimo prazni stavek, smo dobili protislovje.
```

Rezolucijsko pravilo je lokalno pravilo sklepanja, saj za njegovo uporabo zadostujeta dva stavka, ki vsebujeta komplementarna literala. Iz teh dveh stavkov lahko neposredno izpeljemo nov stavek, ne da bi pri tem potrebovali informacije o drugih stavkih v bazi znanja. Zaradi te lastnosti je rezolucija posebej primerna za avtomatsko dokazovanje, saj lahko računalnik sistematično pregleduje pare stavkov in tvori nove rezolvente.

Algoritem {ref}`alg-rezolucija` prikazuje postopek dokazovanja veljavnosti sklepanja z rezolucijo. Pred začetkom zanke za iterativno uporabo rezolucijskega pravila moramo logično formulo sklepanja najprej pretvoriti v konjunktivno normalno obliko (vrstica 9). Spomnimo se, da je vsak konjunkt v konjunktivni normalni obliki stavek. Vsak stavek iz te oblike dodamo v začetno množico za rezolucijski postopek (vrstica 10, množica `K`). Nato uporabljamo rezolucijsko pravilo toliko časa, dokler ne izpeljemo praznega stavka (vrstici 17 in 18) ali ne moremo več izpeljati nobenega novega stavka (vrstici 21 in 22).

```{code-block} text
:caption: Algoritem samodejnega dokazovanja veljavnosti sklepanja z rezolucijo
:name: alg-rezolucija
:linenos:

VHOD:
  A1, A2, ..., An so logične formule iz baze znanja, predpostavke sklepnja 
  B je sklep

IZHOD: True, če je sklepanje veljavno, False, sicer

REZOLUCIJA(A1, A2, ..., An; B)

    F ← (A1 ∧ A2 ∧ ... ∧ An ∧ ¬B)
    K ← množica vseh stavkov v konjunktivni normalni obliki formule F

    while true do
        N ← ∅        # začasna množica z rezolucijo izpeljanih stavkov

        for dvojica stavkov S in T iz K do
            for rezolventa R stavkov S in T do
                if R = □ then
                    return True
                N ← N ∪ {R}

        if N ⊆ K then
            return False

        K ← K ∪ N
```

Algoritem za vsak par stavkov v trenutni množici (zunanja zanka, začetek v vrstici 15) pogleda kolikrat si delita komplementarna literala. Za vsak par komplementarnih literalov (notranja zanka, začetek v vrstici 16) lahko izračunamo rezolvento in jo dodamo množici novih stavkov. Če je rezolventa prazen stavek, smo dosegli protislovje, sicer pa nadaljujemo s preverjanjem rezolvent in parov stavkov. Če po koncu zanke ugotovimo, da v zanki nismo pridobili nobenega novega stavka, lahko ugotovimo, da protislovja ne moremo izpeljati.

Poglejmo sklepanje na osnovi pravila _modus ponens_ iz primera v prejšnjem razdelku. Sklepanje modus ponens lahko zapišemo kot $A, A \implies B \models B$. Preden začnemo z uporabo rezolucijskih korakov, zapišimo logično formulo tega sklepanja s protislovjem $ A \land (A \implies B) \land \neg B $. Prvi in tretji konjunkt v tej formuli sta že stavka (vsak dolžine 1), drugi konjunkt $A \implies B$ pa lahko z znano ekvivalenco iz logike zapišemo v obliko stavka $\neg A \lor B$. Preden začnemo z rezolucijskimi koraki imamo torej tri stavke:

  1. $A$,
  1. $\neg A \lor B$ in
  1. $\neg B$.

Če na stavkih $A$ in $\neg A \lor B$ izvedemo rezolucijski korak glede na literal $A$, po odstranjevanju $A$ in $\neg A$ iz prvega in drugega stavka, dobimo rezolvento $B$, ki jo dodamo naši bazi znanja

  4. $B$.

V naslednjem stavku izpeljemo rezolvento tretjega in četrtega stavka glede na literal $B$. Po odstranjevanju tega literala iz obeh stavkov, dobimo prazni stavek in torej protislovje. S tem smo dokazali, da je modus ponens veljavno pravilo sklepanja.

Za vajo prevedi vsa dejstva, ki so potrebna za sklepanje v detektivski uganki o kraji prototipa iz uvodnega dela tega predavanja, v propozicijsko logiko. Nato uporabi dokazovanje veljavnosti sklepanja z rezolucijo zato, da ugotoviš kdo od osumljencev je ukradel prototip.

Dokažimo zdaj, da je rezolucija korektna (ne more dokazati ničesar, kar ni veljavno) in polna (dokaže vse, kar je veljavno).


```{prf:theorem} Korektnost rezolucijega koraka

Naj bosta $S_1 = L \lor A$ in $S_2 = \neg L \lor B$ stavka, kjer je $L$ literal, $A$ in $B$ pa stavka. Naj bo $R = A \lor B$ rezolventa stavkov $S_1$ in $S_2$. Potem velja $S_1 \land S_2 \models R$.
```

Z drugimi besedami: za vsako določilo, pri katerem sta oba stavka $S_1$ in $S_2$ resnična, je resnična tudi rezolventa.

```{prf:proof} Korektnost rezolucijega koraka

Naj bo $d$ poljubno določilo, pri katerem sta oba stavka $S_1$ in $S_2$ resnična. Potem obravnavamo dve možnosti glede resničnosti literala $L$.

  1. Če ima literal $L$ logično vrednost $\top$, potem je stavek $\neg L \lor B$ resničen zgolj v primeru, če je stavek $B$ resničen, kar pa zagotavlja rezničnost rezolvente $A \lor B$.
  1. Če ima literal $L$ logično vrednost $\bot$, potem je stavek $L \lor A$ resničen zgolj v primeru, če je stavek $A$ resničen, kar pa spet zagotavlja resničnost rezolvente $A \lor B$.
```

```{prf:theorem} Polnost algoritma rezolucija

Naj bo $S$ množica propozicijskih stavkov. Če je konjunkcija stavkov iz $S$ neskladna, potem je mogoče iz stavkov v $S$ z zaporedno uporabo rezolucijskega pravila izpeljati prazni stavek.
```

```{prf:proof} Polnost algoritma rezolucija

Dokaz poteka z indukcijo po številu različnih atomov, ki nastopajo v stavkih iz množice $S$.

**Osnovni korak indukcije** Če v stavkih ne nastopa noben atom, potem so lahko elementa $S$ zgolj trivialna stavka $\top$ in $\bot$. Ker je njihova konjunkcija neskladna, med njimi mora biti $\bot$, ki ustreza praznemu stavku. V tem primeru je torej trditev očitno veljavna.

**Indukcijski korak** Predpostavimo, da izrek velja za vse množice stavkov, v katerih nastopa največ $k$ različnih atomov. Naj bo $S$ množica stavkov, v katerih nastopa $k+1$ atomov. Izberimo enega izmed njih in ga onzačimo z $A$. Zdaj stavke iz množice $S$ razdelimo na tri skupine:

  1. Stavki, ki vsebujejo literal $A$,
  2. Stavki, ki vsebujejo literal $\neg A$ in
  3. Stavki, ki ne vsebujejo atoma $A$.


Iz vsakega para stavkov iz prve in druge skupine $P \lor A$ in $Q \lor \neg A$ tvorimo rezolvento $P \lor Q$. Naj bo $S'$ množica, ki vsebuje vse tako dobljene rezolventa in vse stavke iz tretje skupine. V stavkih iz $S'$ atom $A$ očitno ne nastopa več, torej v $S'$ nastopa največ $k$ atomov.

Pokazati moramo, da je $S'$ neskladna. Predpostavimo nasprotno, da obstaja določilo $d$, za katerega so resnični vsi stavki iz $S'$. To določilo lahko razširimo z vrednostjo za atom $A$ tako, da zadovolji tudi vse stavke iz $S$. Če obstaja stavek oblike $P \lor A$, ki bi bil sicer neresničen, bi nastavili vrednost $A$ na $\top$. Če obstaja stavek oblike $Q \lor \neg A$, ki bi bil sicer neresničen, bi nastavili vrednost $B$ na $\top$. Ker je rezolventa $P \lor Q$ v $S'$, taka izbira vedno obstaja. S tem dobimo razširjeno določilo, ki zadovoljuje vse stavke iz $S$, kar je v nasprotju s predpostavko induktivnega koraka. Torej je $S'$ neskladna.

Množica $S'$ vsebuje manj atomov kot $S$, zato po indukcijski predpostavki iz $S'$ z rezolucijo izpeljemo prazni stavek. Ker so vsi stavki iz $S'$ bodisi stavki iz $S$ bodisi njihove rezolvente, lahko isti dokaz rekonstruiramo z uporabo rezolucije nad stavki iz $S$. Zato iz $S$ z rezolucije izpeljemo prazni stavek $\bot$.
```

Rezolucija je korektna in polna metoda dokazovanja v propozicijski logiki, vendar ima pomembno praktično omejitev. Da lahko rezolucijo uporabimo, moramo vse predpostavke in negacijo sklepa najprej pretvoriti v konjunktivno normalno obliko. Pri tej pretvorbi se lahko dolžina formule zelo poveča: v najslabšem primeru je dolžina konjunktivno normalne oblike formule eksponentna glede na število atomov v izhodiščni formuli. Zato lahko že priprava vhodnih stavkov za rezolucijo postane računsko zelo zahtevna.

Zaradi te težave se v številnih aplikacijah umetne inteligence omejimo na posebne razrede stavkov, za katere je sklepanje učinkovitejše. Posebej pomembni so Hornovi stavki, ki imajo zelo preprosto obliko in omogočajo bistveno hitrejše algoritme sklepanja.

## Hornovi stavki ter veriženje naprej in nazaj

Pri predstavitvi znanja v umetni inteligenci se pogosto pojavljajo pravila oblike če–potem. Takšna pravila opisujejo odvisnosti med dejstvi, na primer:

  1. Če dežuje, je cesta mokra.
  1. Če je cesta mokra in temperatura pod ničlo, je cesta poledenela.
  1. Če ima pacient vročino in kašelj, obstaja možnost okužbe.

Takšna pravila lahko zapišemo v propozicijsko logiko kot logične formule oblike $(A_1 \land A_2 \land ... \land A_n) \implies B$.

V tej formuli so $A_1, A_2, \dots, A_n$ atomi, ki predstavljajo pogoje pravila če-potem, $B$ pa predstavlja njegovo posledico. Na primer, drugo pravilo iz zgornjega seznama lahko prevedemo v propozicijsko logiko tako, da najprej uvedemo atome $A$, cesta je mokra, $B$, temperatura je pod ničlo in $C$, cesta je poledenela. S temi atomi lahko pravilo zapišemo kot $(A \land B) \implies C$. Ta logična formula je ekvivalentna $\neg (A \land B) \lor C$ in naprej še $\neg A \lor \neg B \lor C$, torej stavku z enim pozitivnim literalom. Formalizirajmo to prevedbo logičnih formul pravil če-potem v definicijo Hornovega stavka.

```{prf:definition} Hornerov stavek

Hornov stavek je stavek, ki vsebuje največ en pozitiven literal.
```

Hornovi stavki so treh vrst:

| Vrsta | Oblika | Pomen (interpretacija) |
|:------|:-------|:---------------|
| Dejstvo | $B$ | $B$ velja. |
| Pravilo | $\neg A_1 \land \neg A_2 \land \dots \land \neg A_n \land B$ | Če veljajo pogoji $A_1, A_2, \dots, A_n$, velja sklep |
| Omejitev | $\neg A_1 \land \neg A_2 \land \dots \land \neg A_n$ | Pogoji $A_1, A_2, \dots, A_n$ ne smejo veljati hkrati |

Hornovi stavki omogočajo zelo naraven način sklepanja. Če že poznamo nekaj dejstev, lahko z uporabo pravil postopoma izpeljujemo nova dejstva. Temu postopku pravimo veriženje naprej (angl. _forward chaining_). Pri tem začnemo z znanimi dejstvi in nato večkrat uporabimo pravila, katerih pogoji so izpolnjeni, dokler ne dobimo novih sklepov.

Sklepanje lahko poteka tudi v obratni smeri. Če želimo preveriti, ali velja podan cilj $B$, poiščemo pravilo, katerega posledica je $B$, in nato poskušamo dokazati njegove pogoje $A_1, A_2, \dots, A_n$. Temu postopku pravimo veriženje nazaj (angl. backward chaining). Pri tem začnemo s ciljem in se postopoma vračamo k dejstvom, ki jih že poznamo.

Poglejmo primer. Naj bo podana baza znanja z atomi $A$, dežuje, $B$, cesta je mokra, $C$, cesta je spolzka, $D$, temperatura je pod ničlo in $E$, obstaja nevarnost nesreče in Hornovimi stavki

  1. $A$
  1. $A \implies B$
  1. $B \implies C$
  1. $C \land D \implies E$
  1. $D$

Pri veriženju naprej (algoritem {ref}`alg-verizenje`) začnemo z dejstvi v bazi znanja, torej $A$ in $D$. Iz dejstva $A$ z uporabo drugega pravila, izpeljemo dejstvo $B$. Iz tega dejstva z uporabo tretjega pravila, izpeljemo dejstvo $C$. Iz tega dejstva in dejstva $D$ z uporabo četrtega pravila, izpeljemo dejstvo $E$.

```{code-block} text
:caption: Algoritma dokazovanja z veriženjem naprej in nazaj
:name: alg-verizenje
:linenos:

VERIZENJE_NAPREJ(KB)

    F ← množica znanih dejstev iz KB

    repeat
        changed ← false

        for pravilo (A1 ∧ ... ∧ An → B) v KB do
            if A1,...,An ∈ F in B ∉ F then
                dodaj B v F
                changed ← true
    until changed = false

    return F


VERIZENJE_NAZAJ(KB, cilj)

    if cilj je znano dejstvo then
        return true

    for pravilo (A1 ∧ ... ∧ An → goal) v KB do
        if VERIZENJE NAZAJ(KB, A1) in ... in VERIZENJE_NAZAJ(KB, An) then
            return true

    return false
```

Recimo, da z veriženjem nazaj želimo preveriti dejstvo $E$. To dejstvo je posledica četrtega pravila, katerega pogoj je $C \land D$. Torej zdaj moramo dokazati $C$ in $D$. Dejstvo $C$ je posledica tretjega pravila, katerega pogoj je $B$. Torej moramo dokazati $B$, ki je posledica drugega pravila s pogojem $A$. Ker je $A$ znano dejstvo v bazi znanja, smo dejstvo $C$ dokazali. Dejstvo $D$ je znano dejstvo v bazi znanja. Ker smo dokazali vse pogoje, sledi $E$, dejstvo, katerega veljavnost smo hoteli preveriti.

Torej, na kratko. Veriženje naprej začne z znanimi dejstvi in iz njih postopoma izpeljuje nove sklepe. Veriženju nazaj pa začne z izbranim ciljnim dejstvom in poskuša dokazati pogoje pravil, ki vodijo do tega ciljnega dejstva.

Rezolucija je splošna metoda dokazovanja v propozicijski logiki in je, kot smo videli, korektna in polna. Vendar ima lahko v najslabšem primeru zelo visoko časovno zahtevnost. Število možnih rezolvent lahko hitro naraste, zato je dokazovanje z rezolucijo v splošnem primeru **eksponentno zahtevno** glede na velikost baze znanja.

Pri Hornovih stavkih pa je položaj bistveno bolj ugoden. Zaradi njihove posebne strukture lahko sklepanje izvajamo z algoritmoma veriženja naprej ali veriženja nazaj, ki sta bistveno učinkovitejša. Za množico Hornovih stavkov lahko z veriženjem naprej ugotovimo, ali nek atom logično sledi iz baze znanja v času **linearno odvisnem od velikosti baze znanja**. Naj bo namreč baza znanja sestavljena iz $m$ pravil in $n$ atomov. Pri veriženju naprej lahko vsak atom dodamo med znana dejstva največ enkrat, vsako pravilo pa tudi lahko uporabimo največ enkrat. Zato je število korakov veriženja omejeno z $O(m + n)$. Z veriženjem nazaj pa lahko preverimo veljavnost ciljnega dejstva v času **linearno odvisnem od velikosti relevantnega dela baze znanja**. To pomeni, da je sklepanje nad Hornovimi stavki bistveno učinkovitejše kot splošno rezolucijsko dokazovanje. Zato se Hornovi stavki pogosto uporabljajo v sistemih umetne inteligence za predstavitev znanja, na primer v logičnem programiranju in v jeziku Prolog.
