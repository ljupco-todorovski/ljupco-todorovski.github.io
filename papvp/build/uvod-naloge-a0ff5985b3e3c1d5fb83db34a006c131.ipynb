# Naloge za vaje

Naloge utrjujejo izpeljane sezname, izpeljane množice in funkcije višjega reda. Rešitve preveri s podanimi primeri. Zadostujejo osnovne operacije v Pythonu; dodatnih knjižnic ne potrebuješ.

## Izpeljani seznami: pretvorba časov

V seznamu `trajanja` so trajanja posnetkov v sekundah. Z enim izpeljanim seznamom sestavi seznam `minute`, ki vsebuje ista trajanja v minutah. Ohrani vrstni red posnetkov.

```python
trajanja = [90, 120, 45, 180]
```

Za podane podatke pričakujemo rezultat:

```python
[1.5, 2.0, 0.75, 3.0]
```

Preveri še, da za prazen seznam trajanj dobiš prazen seznam minut.

## Izpeljani seznami: urejanje in izbira imen

Seznam `imena` vsebuje imena z morebitnimi odvečnimi presledki na začetku ali koncu. Z enim izpeljanim seznamom sestavi seznam `izbrana_imena`, ki vsebuje samo imena z **vsaj štirimi znaki po odstranitvi presledkov**. V rezultatu naj bodo imena brez odvečnih presledkov, njihov vrstni red pa naj ostane enak.

```python
imena = [" Ana ", "Boris", " Eva", "  Maja ", "Tim", "  "]
```

Za podane podatke pričakujemo rezultat:

```python
["Boris", "Maja"]
```

Namig: metoda `ime.strip()` odstrani presledke na začetku in koncu niza, `len(ime)` pa vrne njegovo dolžino. Presledke odstrani tako pri preverjanju pogoja kot pri tvorjenju rezultata.

## Izpeljane množice: različna mesta

V seznamu `mesta` so nekatera mesta zapisana večkrat, z različnimi velikimi in malimi črkami ter odvečnimi presledki. Z izpeljano množico sestavi množico `razlicna_mesta`, v kateri je vsako mesto zapisano samo enkrat, z malimi črkami in brez presledkov na začetku ali koncu.

```python
mesta = ["Ljubljana", " Maribor ", "ljubljana", "Koper", "MARIBOR", " Koper "]
```

Za podane podatke pričakujemo množico z elementi `"ljubljana"`, `"maribor"` in `"koper"`. Množica ne zagotavlja vrstnega reda; za urejen izpis uporabi `sorted(razlicna_mesta)`. Pričakovani izpis je:

```python
["koper", "ljubljana", "maribor"]
```

Namig: metodi lahko povežeš v izraz `mesto.strip().lower()`.

## Funkcije višjega reda: funkcija kot argument

Definiraj funkcijo `vsota_po(f, podatki)`, ki sprejme funkcijo `f` in seznam `podatki`. Za vsak element seznama naj izračuna vrednost `f(element)`, nato pa vrne vsoto teh vrednosti. Za prazen seznam naj vrne `0`.

Preveri naslednje klice:

```python
vsota_po(lambda x: x ** 2, [-2, 1, 3])  # 14
vsota_po(abs, [-2, 1, 3])              # 6
vsota_po(lambda x: x ** 2, [])          # 0
```

Namig: uporabi izpeljani seznam in funkcijo `sum`. Funkcijo `f` podaj kot argument in jo pokliči na posameznih elementih znotraj funkcije `vsota_po`.

## Funkcije višjega reda: funkcija kot rezultat

Definiraj funkcijo `ustvari_linearno_funkcijo(a, b)`, ki sprejme števili `a` in `b` ter **vrne novo funkcijo**. Vrnjena funkcija naj sprejme število `x` in vrne vrednost `a * x + b`.

Preveri naslednji primer:

```python
dvojno_plus_ena = ustvari_linearno_funkcijo(2, 1)
nasprotno = ustvari_linearno_funkcijo(-1, 0)

print(dvojno_plus_ena(3))  # 7
print(nasprotno(3))       # -3
print(dvojno_plus_ena(0)) # 1
```

Obe ustvarjeni funkciji morata delovati tudi po tem, ko se klic `ustvari_linearno_funkcijo` konča. Preveri, da druga ustvarjena funkcija ne spremeni delovanja prve.

Namig: znotraj `ustvari_linearno_funkcijo` definiraj funkcijo z argumentom `x`. Ta lahko uporablja argumenta `a` in `b` zunanje funkcije. Na koncu vrni notranjo funkcijo.
