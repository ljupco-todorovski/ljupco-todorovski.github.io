# Izpeljani seznami (angl. _list comprehensions_)

Izpeljani seznami v programskem jeziku Python omogočajo jedrnato ustvarjanje novih seznamov. Osnovna oblika izpeljanega seznama je `[izraz for e in vir if pogoj]`:

  * `vir` je iterabilni objekt (angl. _iterable_), na primer seznam, niz, objekt `range` ali iterator;
  * `e` je spremenljivka, ki predstavlja trenutni element vira;
  * `izraz` je izraz v Pythonu, ki na podlagi vrednosti `e` izračuna element novega seznama;
  * `pogoj` je izraz, katerega logična vrednost določa, ali bomo trenutni element vira uporabili pri tvorjenju novega seznama.

Del `if pogoj` ni obvezen. Brez njega dobimo obliko `[izraz for e in vir]`, ki za vsak element vira ustvari en element novega seznama. Pri tem predpostavljamo, da je vir končen. Izpeljani seznam se v celoti ustvari takoj, pri prehodu skozi iterator pa se njegovi elementi porabijo.

Pri podatkovni analizi pogosto obravnavamo podatke, shranjene v tabelah, računamo z obstoječimi stolpci in ustvarjamo nove. Izpeljani seznami so pri tem lahko uporabni, čeprav knjižnica `pandas` ponuja tudi svoje operacije za delo s stolpci. Razumevanje izpeljanih seznamov ter njihove povezave s funkcijama višjega reda `map` in `filter` je zato koristna priprava na podatkovno analizo s Pythonom.

## Preprost primer in alternativi

Oglejmo si izpeljani seznam kvadratov celih števil od 0 do 9. Klic `range(10)` vključuje 0, ne vključuje pa 10:

```python
kvadrati = [e ** 2 for e in range(10)]
print(kvadrati)
```

Isti seznam `kvadrati` lahko ustvarimo tudi drugače. Prvi alternativni način uporablja zanko `for`:

```python
kvadrati_a = []
for e in range(10):
    kvadrati_a.append(e ** 2)
print(kvadrati_a)
```

Drugi način uporablja funkcijo `map(f, vir)`, ki za vsak element vira uporabi funkcijo `f`. Funkcija `map` vrne iterator, zato si najprej oglejmo tip rezultata:

```python
kvadrati_b = map(lambda e: e ** 2, range(10))
print(type(kvadrati_b))
```

Objekt tipa `map` še ni seznam: vrednosti izračunava sproti, ko jih zahtevamo. S funkcijo `list` lahko iterator porabimo in njegove rezultate zberemo v seznam:

```python
kvadrati_b = list(kvadrati_b)
print(kvadrati_b)
```

V tem primeru je izpeljani seznam krajši in preglednejši od zanke z metodo `append` ali kombinacije funkcij `list`, `map` in `lambda`. Če pa že imamo ustrezno funkcijo, je lahko tudi uporaba `map` jedrnata in razumljiva.

## Primer s pogojem

Ustvarimo zdaj seznam kvadratov lihih celih števil med 0 in 9:

```python
kvadrati_lihi = [e ** 2 for e in range(10) if e % 2 == 1]
print(kvadrati_lihi)
```

Za vajo sestavi program, ki z zanko ustvari isti seznam in ga shrani v spremenljivko `kvadrati_lihi_a`.

Tukaj bomo uporabili kombinacijo funkcij `map` in `filter`. Funkcija `filter` vrne iterator z elementi vira, za katere ima podani pogoj logično vrednost `True`. S funkcijo `list` jih zberemo v seznam:

```python
list(filter(lambda e: e % 2 == 1, range(10)))
```

S funkcijo `filter` najprej izberemo liha števila, nato pa s funkcijo `map` izračunamo njihove kvadrate:

```python
kvadrati_lihi_b = list(map(lambda e: e ** 2, filter(lambda e: e % 2 == 1, range(10))))
print(kvadrati_lihi_b)
```

## Gnezdenje izpeljanih seznamov

Izpeljane sezname lahko gnezdimo: izraz v zunanjem izpeljanem seznamu je lahko nov izpeljani seznam. Tako lahko matriko predstavimo kot seznam vrstic, pri čemer je vsaka vrstica seznam elementov. Ustvarimo na primer enotsko matriko velikosti 3 × 3, ki ima na glavni diagonali enice, drugod pa ničle:

```python
im_3 = [[1 if i == j else 0 for j in range(3)] for i in range(3)]
print(im_3)
```

V tem primeru je `i` indeks vrstice, `j` pa indeks stolpca. Pogojni izraz `1 if i == j else 0` določa, **katero vrednost** dodamo v seznam. Nasprotno pa `if pogoj` za delom `for` določa, **ali element sploh dodamo**. Pogojni izraz torej izbira med dvema vrednostma, pogoj na koncu izpeljanega seznama pa filtrira elemente vira.

En izpeljani seznam lahko vsebuje tudi več delov `for`, ne da bi s tem ustvaril seznam seznamov. Enotsko matriko lahko na primer sploščimo v en seznam:

```python
[element for vrstica in im_3 for element in vrstica]
```

Za vajo primerjaj rezultata zadnjih dveh primerov kode in pojasni razliko. Dele `for` beremo od leve proti desni, tako kot pri gnezdenih zankah: najprej izberemo vrstico, nato pa njene elemente.

Zdaj, ko vemo, kako gnezdimo izpeljane sezname, definirajmo funkcijo `diag`, ki ustvari diagonalno matriko iz podanega seznama elementov na glavni diagonali:

```python
def diag(d):
    n = len(d)
    return [[d[i] if i == j else 0 for j in range(n)] for i in range(n)]


print(diag([-2, -1, 0, 1, 2]))
```

Za vajo napiši še dve različici funkcije `diag`: ena naj uporablja zanke, druga pa funkcijo `map`. Primerjaj njuno dolžino in berljivost z zgornjo različico. Če želiš primerjati tudi čas izvajanja, ga izmeri pri različnih velikostih vhodnega seznama; krajša koda ni nujno hitrejša.

## Kombiniranje seznamov

Izpeljani seznam lahko združuje podatke iz več virov. Za seštevanje istoležnih elementov dveh seznamov lahko uporabimo skupni indeks ali funkcijo `zip`, ki združi istoležne elemente v pare. Funkcija `plus` uporabi drugi pristop in preveri, ali sta seznama enako dolga:

```python
def plus(x, y):
    if len(x) != len(y):
        raise ValueError("Seznama morata biti enako dolga.")
    return [a + b for a, b in zip(x, y)]


plus([-1, 0, 1], [1, 0, -1])
```

Klic `zip(x, y)` se privzeto ustavi, ko se izčrpa krajši vir. Zato pred seštevanjem preverimo dolžini in pri različnih dolžinah sprožimo izjemo `ValueError`.

Za vajo posploši funkcijo `plus`, da sprejme poljubno število seznamov in sešteje njihove istoležne elemente. Preveri, da so vsi seznami enako dolgi, in določi, kaj naj funkcija vrne, če ne podamo nobenega seznama. Namig: uporabi parameter `*seznami`, klic `zip(*seznami)` in funkcijo `sum`.

## Prednosti in slabosti izpeljanih seznamov

Prednosti:

  * Pri preprostih preslikavah in filtriranju omogočajo jedrnato in berljivo kodo.
  * V primerjavi z zanko, ki seznam gradi s klici metode `append`, so pogosto hitrejši; hitrost je odvisna od konkretnega izračuna in jo preverimo z meritvami.

Slabosti:

  * Veliko pogojev, gnezdenih zank ali zapleten izraz lahko zmanjšajo berljivost. V takih primerih raje uporabimo zanke ali pomožne funkcije.
  * Celoten seznam ustvarijo takoj in ga shranijo v pomnilnik. Če rezultate potrebujemo le sproti, je lahko primernejši generatorski izraz `(izraz for e in vir if pogoj)`.
  * Za zaporedne izračune, pri katerih novi rezultat temelji na prejšnjem, so običajno preglednejše zanke ali namenske funkcije, na primer `itertools.accumulate` za kumulativne izračune.

## Izpeljane množice

Če v obliki `[izraz for e in vir if pogoj]` oglate oklepaje zamenjamo z zavitimi, dobimo izpeljano množico `{izraz for e in vir if pogoj}`. Množica vsebuje le različne elemente in ne zagotavlja vrstnega reda. Njeni elementi morajo biti zgoščljivi (angl. _hashable_); nizi so, seznami pa niso.

Iz seznama e-poštnih naslovov izluščimo množico različnih domen. Domena e-poštnega naslova je del za znakom `@`: pri naslovu `ljupco.todorovski@fmf.uni-lj.si` je to `fmf.uni-lj.si`.

Najprej si oglejmo metodo `split`, ki niz razdeli na seznam podnizov glede na podano ločilo:

```python
print("ljupco.todorovski@fmf.uni-lj.si".split("@"))
```

Funkcijo za pridobivanje domene definiramo takole. Predpostavljamo, da dobimo veljaven e-poštni naslov v preprosti obliki `ime@domena`; funkcija veljavnosti naslova ne preverja:

```python
def domena(naslov):
    return naslov.split("@")[-1]

print(domena("ljupco.todorovski@fmf.uni-lj.si"))
```

Funkcijo uporabimo v izpeljani množici. Za ponovljiv vrstni red izpisa domene uredimo s funkcijo `sorted`, ki vrne seznam:

```python
naslovi = [
    "ljupco.todorovski@fmf.uni-lj.si",
    "ajda.lampe@fmf.uni-lj.si",
    "ajda.lampe@fri.uni-lj.si",
    "Katja.StembergerBrizani@pf.uni-lj.si",
    "ljupco@gmail.com",
    "nekdo_drug@gmail.com",
]

domene = {domena(n) for n in naslovi}
print(sorted(domene))
```

Za vajo prilagodi zgornjo kodo tako, da izpiše urejen seznam različnih **vrhnjih domen** podanih e-poštnih naslovov. Vrhnja domena (angl. _top-level domain_, TLD) je zadnji del domene, za zadnjo piko `.`. Tako je na primer `si` vrhnja domena v domeni `fmf.uni-lj.si`.

V [uradni dokumentaciji Pythona](https://docs.python.org/3/tutorial/datastructures.html#dictionaries) poišči še opis _izpeljanih slovarjev_ (angl. _dictionary comprehensions_). Nato sestavi slovar, ki vsak naslov iz seznama `naslovi` preslika v njegovo domeno.
