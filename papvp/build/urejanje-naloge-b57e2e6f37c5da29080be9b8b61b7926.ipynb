# Naloge za vaje

Naloge utrjujejo uvoz in urejanje podatkov, združevanje tabel ter delo z besedilom in regularnimi izrazi. Uporabi knjižnico `pandas`, uvoženo kot `pd`. Pri nalogah s tabelama `studenti` in `studenti_ocene` najprej izvedi ustrezne primere iz zvezka [Urejanje podatkov in združevanje tabel](urejanje-uvoz-sql.ipynb). Tabelo imen in kratic članic univerze pripravi po zvezku [Urejanje besedilnih podatkov in regularni izrazi](urejanje-uvoz-ri.ipynb).

## Urejanje in združevanje tabel

1. **Indeks tabele študentov.** Indeks urejene tabele `studenti` spremeni na dva načina. Najprej naj bodo oznake vrstic zaporedna cela števila od 0 do 5 brez ponovitev. Nato uporabi imena študentov kot oznake vrstic. Različici pripravi neodvisno iz iste izvirne tabele.

1. **Tipi in domene spremenljivk.** V tabeli `studenti_ocene` pretvori stolpca `semester` in `kolokvij` v kategorični tip. Izračunaj domeno (množico ali seznam različnih vrednosti) vsake spremenljivke. Loči manjkajoče podatke od vrednosti v domeni.

1. **Število kombinacij.** Pojasni, zakaj ima kartezični produkt domen dimenzijskih spremenljivk tabele `studenti_ocene` 24 vrstic. Število preveri iz velikosti domen in ga primerjaj s številom vrstic dopolnjene tabele.

1. **Izidi terapij.** V datoteki [`izidi.xlsx`](https://kt.ijs.si/~ljupco/lectures/papvp-2324/izidi.xlsx) so podatki o številskih izidih dveh terapij pri treh pacientih. Prazne celice pomenijo manjkajoče rezultate. Napiši program, ki prebere preglednico in sestavi urejeno podatkovno tabelo.

    Namig: tabela naj ima dve dimenzijski spremenljivki in eno merjeno spremenljivko. Pri preoblikovanju ohrani tudi zapise z manjkajočimi rezultati.

1. **Kavni napitki v bistroju Mafija.** V bistroju Mafija strežejo črno kavo, kavo z mlekom in kavo s smetano. Ob letni bilanci zapišejo število postreženih skodelic posameznega napitka:

    | Vrsta kavnega napitka | 2018 | 2019 | 2020 | 2021 | 2022 |
    |:---|---:|---:|---:|---:|---:|
    | Črna kava | 23 100 | 3 500 | 6 200 | 18 700 | 21 400 |
    | Kava z mlekom | 18 300 | 2 800 | 4 900 | 14 000 | 18 000 |
    | Kava s smetano | 11 500 | 1 600 | 3 400 | 7 600 | 11 100 |

    Premisli, katere spremenljivke opisujejo opazovano enoto. Podatke preoblikuj v urejeno tabelo v knjižnici `pandas`.

1. **Selitveno gibanje.** Statistični urad Republike Slovenije v podatkovni bazi [SiStat](https://pxweb.stat.si/) objavlja podatke o selitvenem gibanju prebivalstva med regijami in tujino. Izvoz teh podatkov je v datoteki [`05I2004S_20221025-231704.xlsx`](https://kt.ijs.si/~ljupco/lectures/papvp-2324/05I2004S_20221025-231704.xlsx). Napiši program, ki iz preglednice sestavi urejeno podatkovno tabelo. Pri uvozu loči podatke od naslovov, opomb in drugih spremljevalnih besedil.

1. **Potrjene okužbe po regijah.** Projekt [COVID-19 Sledilnik](https://covid-19.sledilnik.org/) hrani podatke o epidemiji v Sloveniji. Zgodovinski dnevni podatki so v [repozitoriju GitHub](https://github.com/sledilnik/data/blob/master/csv/stats.csv) in neposredno v [datoteki CSV](https://raw.githubusercontent.com/sledilnik/data/master/csv/stats.csv).

    Iz podatkov sestavi urejeno tabelo s tremi stolpci:

    * `datum`: dimenzijska spremenljivka z datumom opazovanja;
    * `regija`: dimenzijska spremenljivka z imenom ene od dvanajstih [statističnih regij Slovenije](https://sl.wikipedia.org/wiki/Statisti%C4%8Dne_regije_Slovenije) ali vrednostjo za Slovenijo kot celoto;
    * `kumulativne_okuzbe`: merjena spremenljivka s kumulativnim številom potrjenih okužb do izbranega datuma.

    Preveri pomen regionalnih oznak in stolpcev v viru. Manjkajočih vrednosti ne nadomeščaj samodejno z ničlami. Podatki so zgodovinski; ne predpostavi, da se še dnevno osvežujejo.

1. **Okužbe glede na število prebivalcev.** V bazi [SiStat](https://pxweb.stat.si/) poišči število prebivalcev po statističnih regijah in za Slovenijo po polletjih. Izvozi vse razpoložljive podatke od prvega polletja 2020 v preglednico Excel in jih uredi v tabelo s stolpci `polletje`, `regija` in `prebivalstvo`.

    Tabelo poveži s podatki iz prejšnje naloge ter izračunaj razmerje med kumulativnim številom potrjenih okužb in številom prebivalcev za vsako regijo in Slovenijo. Dnevom pripiši ustrezno polletje in računaj za obdobje, v katerem sta na voljo oba podatka. Razmerje samo po sebi ni delež različnih okuženih oseb, saj je bila ista oseba lahko okužena večkrat.

## Besedilni podatki in regularni izrazi

1. **Vaje z regularnimi izrazi.** Rešuj naloge in križanke na straneh [RegexOne](https://regexone.com/), [Regex Crossword](https://regexcrossword.com/) in [Regex Golf](https://alf.nu/RegexGolf). Upoštevaj, da se podprta sintaksa lahko razlikuje od sintakse modula `re` v Pythonu.

1. **Naslovi e-pošte.** Iz datoteke [`enaslovi.txt`](https://kt.ijs.si/~ljupco/lectures/papvp-2324/enaslovi.txt) preberi naslove osmih namišljenih sodelavcev Univerze v Ljubljani. Sestavi podatkovno tabelo s stolpci `ime`, `priimek`, `fakulteta` in `enaslov`.

    Naslovi sledijo vzorcu `ime.priimek@kratica.uni-lj.si`. Dela `ime` in `priimek` določata ime in priimek osebe, `kratica` pa članico univerze, na kateri je zaposlena. Pomagaj si s tabelo imen in kratic članic iz predavanj. Preveri, ali imajo vse kratice iz datoteke ustrezen zapis v tabeli.

1. **Samoglasniki in soglasniki.** Definiraj funkcijo `n_samoglasnikov`, ki za podan niz vrne število samoglasnikov. Nato definiraj funkcijo `n_soglasnikov`, ki prešteje soglasnike. Upoštevaj velike in male črke ter slovenske šumnike. Presledki, števke in ločila niso soglasniki.

1. **Besede v besedilu.** Datoteka [`besedilo.txt`](https://kt.ijs.si/~ljupco/lectures/papvp-2324/besedilo.txt) vsebuje besedilo iz slovenske Wikipedije. S funkcijama iz prejšnje naloge sestavi tabelo `besede` s tremi stolpci:

    * `beseda`: dimenzijska spremenljivka, ki vsebuje vsako različno besedo iz besedila enkrat;
    * `n_samoglasnikov`: število samoglasnikov v besedi;
    * `n_soglasnikov`: število soglasnikov v besedi.

    Določi, kako obravnavaš velike in male črke, ter preveri, da v besedah ni ločil.

1. **Občine in regije.** V članku [Statistične regije Slovenije](https://sl.wikipedia.org/wiki/Statisti%C4%8Dne_regije_Slovenije) razdelek _Občine po statističnih regijah_ vsebuje sezname občin za vseh dvanajst regij. Napiši program, ki iz teh seznamov sestavi urejeno tabelo `obcina_regija` z dimenzijsko spremenljivko `obcina` in merjeno spremenljivko `regija`.

1. **Prebivalstvo in površina regij.** Iz istega članka uvozi tabelo s številom prebivalcev in površino v kvadratnih kilometrih za vseh dvanajst statističnih regij. Preveri naslov in vsebino tabele, saj se razdelki strani spreminjajo. Zabeleži tudi referenčno leto podatkov in preveri podatkovne tipe številskih stolpcev.

1. **Bruto domači proizvod držav.** Dokončaj uvoz podatkov o nominalnem bruto domačem proizvodu iz [članka v Wikipediji](https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)). Sestavi tabelo s stolpci `drzava`, `celina`, `leto` in `bdp`. Izberi en vir ocen BDP in zabeleži enoto vrednosti.

    Odstrani oznake opomb, ustrezno obravnavaj manjkajoče vrednosti in nastavi podatkovne tipe stolpcev. Letnico preveri v podatkih oziroma glavi tabele. Če trenutna različica tabele nima podatka o celini, ga dodaj iz drugega vira in ta vir navedi. Upoštevaj, da geografske regije OZN niso povsem enake celinam. Skupnih vrednosti za svet ne obravnavaj kot države.
