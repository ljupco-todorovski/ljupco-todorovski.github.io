# Naloge za vaje

Naloge utrjujejo ustvarjanje in indeksiranje zaporedij, kategorične in časovne podatkovne tipe ter delo s podatkovnimi tabelami. Uporabi knjižnico `pandas`, uvoženo kot `pd`; za naključne podatke si lahko pomagaš z modulom `random`, za matematične funkcije pa s knjižnico `numpy`, uvoženo kot `np`. Za ponovljive rezultate nastavi seme generatorja, ki ga uporabljaš. Rešitve preveri tudi na mejnih primerih.

1. **Indeks telesne mase.** Alešu, Barbari, Cirilu in Darji smo izmerili višine v centimetrih (180, 165, 160, 193) in mase v kilogramih (87, 58, 65, 100). Ustvari zaporedji `v` in `t`, katerih oznake so imena oseb v podanem vrstnem redu. Indeks telesne mase (ITM) izračunamo po formuli

    $$\text{ITM} = \frac{\text{masa v kilogramih}}{(\text{višina v metrih})^2}.$$

    Izračunaj zaporedje `itm` z vrednostmi ITM in istimi oznakami. Nato izračunaj zaporedje naravnih logaritmov teh vrednosti. Izpiši seznam imen oseb z ITM nad 25 in izračunaj njihov povprečni ITM. Za Aleša mora biti ITM približno 26,85.

    Namig: višine pred izračunom pretvori v metre. Za naravni logaritem uporabi `np.log`, za izbiro oseb pa logični pogoj.

1. **Ustvarjanje zaporedij.** Sestavi tri zaporedja z naslednjimi elementi:

    * $3^1/1, 3^2/2, \ldots, 3^{50}/50$;
    * nizi `"A1"`, `"A2"`, ..., `"A50"`;
    * vrednosti $e^x \sin x$ v 31 točkah $x = 3, 3.1, 3.2, \ldots, 6$.

    Preveri dolžine zaporedij in njihove prve ter zadnje elemente. Namig: pri prvem zaporedju računaj potence z običajnimi celimi števili v Pythonu, saj $3^{50}$ preseže obseg tipa `int64`. Za tretje zaporedje lahko točke ustvariš z `np.linspace(3, 6, 31)`.

1. **Preverjanje praštevil.** Definiraj funkcijo, ki sprejme celo število `x > 2` in vrne `True`, če je `x` praštevilo, sicer pa `False`. Uporabi zaporedje ostankov deljenja `x` s števili od 2 do `x - 1`.

    Namig: preveri, ali je kateri od ostankov enak 0, z izrazom `(ostanki == 0).any()`. Izraz `0 in ostanki` pri zaporedju preverja prisotnost **oznake v indeksu**, ne vrednosti. Preveri, da funkcija za 3 in 17 vrne `True`, za 4 in 9 pa `False`.

1. **Računanje z zamaknjenimi zaporedji.** Ustvari zaporedji `x` in `y` dolžine 250. Vsak element naj bo neodvisno naključno izbrano celo število od 0 do 999, vključno z obema mejama; vrednosti se lahko ponavljajo. Označimo elemente z $x_1, \ldots, x_{250}$ in $y_1, \ldots, y_{250}$. Izračunaj:

    * zaporedje $y_2 - x_1, y_3 - x_2, \ldots, y_{250} - x_{249}$ z oznakami od 0 do 248;
    * vsoto $\sum_{i=1}^{249} e^{-x_{i+1}} / (x_i + 10)$.

    Namig: aritmetične operacije poravnajo zaporedja po oznakah, zato samo rezanje z `iloc` ne zadostuje za želeni zamik. Pred računanjem ustrezno nastavi indekse ali uporabi `shift`. Pri prvem izrazu preveri, da dobiš 249 elementov brez manjkajočih vrednosti.

1. **Kategorično zaporedje.** Ustvari kategorično zaporedje s kategorijami `A`, `B` in `C`, ki vsebuje natanko pet elementov vsake kategorije v naključnem vrstnem redu. Preveri dolžino 15 in število pojavitev posamezne kategorije. Namig: naključno premešaj seznam, ki že vsebuje pet ponovitev vsake vrednosti.

1. **Pretvorba dosežkov v ocene.** Za potrebe te naloge uporabimo naslednjo lestvico. Dosežek je število odstotnih točk od 0 do 100, vključno z obema mejama:

    | Dosežek | Ocena |
    |:---|:---|
    | Več kot 90 | odlično (10) |
    | Več kot 80 in največ 90 | prav dobro (9) |
    | Več kot 70 in največ 80 | prav dobro (8) |
    | Več kot 60 in največ 70 | dobro (7) |
    | Več kot 50 in največ 60 | zadostno (6) |
    | Največ 50 | nezadostno (5) |

    * Ustvari zaporedje `ot` z desetimi naključnimi celoštevilskimi dosežki med 0 in 100.
    * S funkcijo `pd.cut` ustvari urejeno kategorično zaporedje `ocena` z zgornjimi poimenovanji, urejenimi od ocene 5 do ocene 10.

    Preveri še razvrstitev dosežkov 0, 50, 51, 60, 70, 80, 90 in 100. Nobeden od njih ne sme dobiti manjkajoče vrednosti. Namig: upoštevaj vključevanje desnih mej in poskrbi, da je vključena tudi spodnja meja 0.

1. **Dnevi v tednu.** Definiraj funkcijo, ki za podano leto vrne zaporedje datumov vseh ponedeljkov v tem koledarskem letu. Nato jo posploši tako, da sprejme še dan v tednu kot celo število od 0 (ponedeljek) do 6 (nedelja). Preveri, da vsi datumi pripadajo zahtevanemu letu in dnevu v tednu.

1. **Tedni v letu.** Definiraj funkcijo, ki za podano leto in števili tednov `od` in `do` vrne zaporedje vseh datumov v tednih od `od` do `do`, vključno z obema tednoma. Uporabi **tedensko leto po standardu ISO**: teden se začne v ponedeljek, prvi teden pa je tisti, ki vsebuje 4. januar. Tedensko leto ima 52 ali 53 tednov; lahko se začne v prejšnjem koledarskem letu in konča v naslednjem.

    Zahtevaj `1 <= od <= do` in preveri, ali oba tedna v podanem tedenskem letu obstajata. Namig: uporabi `date.fromisocalendar` iz modula `datetime`. Za leto 2020 in prvi teden mora rezultat vsebovati datume od 30. decembra 2019 do 5. januarja 2020.

1. **Naključni datumi.** Definiraj funkcijo, ki sprejme začetni datum `od`, končni datum `do` in nenegativno celo število `n`. Vrne naj zaporedje `n` naključno izbranih datumov med obema mejama, vključno z njima. Predpostavi datuma brez časovnega dela in zahtevaj `od <= do`.

    Datume izbiraj s ponavljanjem, saj ima obdobje lahko manj kot `n` dni. Preveri primere `n = 0`, `od == do` in obdobje, ki vsebuje prestopni dan. Pri `od == do` morajo biti vsi izbrani datumi enaki podanemu datumu.

1. **Podatkovna tabela oseb.** Iz podatkov prve naloge ustvari tabelo s stolpci `ime`, `visina`, `teza` in `ITM`. Stolpec `teza` naj vsebuje mase v kilogramih, stolpec `visina` pa višine v centimetrih. Nato uporabi imena oseb kot oznake vrstic in odstrani stolpec `ime`. Z logičnim indeksiranjem izpiši seznam imen oseb z ITM največ 25.

    Namig: pri sestavljanju tabele upoštevaj poravnavo zaporedij po oznakah. Za nastavitev indeksa preuči metodo `set_index`.

1. **Temperature ob petkih trinajstega.** Preberi tabelo povprečnih dnevnih temperatur zraka v Ljubljani od začetka leta 2010 do konca leta 2019:

    ```python
    temperature = pd.read_csv(
        "https://kt.ijs.si/~ljupco/lectures/papvp-2324/temp-lj-2010-2019.csv",
        parse_dates=["datum"]
    )
    ```

    Tabela ima stolpca `datum` in `povprecna_temperatura`; temperature so v stopinjah Celzija. Če uporabljaš lokalno kopijo `temp-lj-2010-2019.csv` v korenu repozitorija, URL zamenjaj s potjo do te datoteke glede na delovno mapo.

    * Ustvari zaporedje `temp_lj` z datumi kot oznakami in temperaturami kot vrednostmi. Indeks uredi po datumih.
    * Z indeksiranjem ustvari zaporedje `temp_lj_petek` s temperaturami ob petkih.
    * Z indeksiranjem ustvari zaporedje `temp_lj_13` s temperaturami na 13. dan v mesecu.
    * Z metodo `align` in ustreznim argumentom `join` poravnaj ti zaporedji tako, da ostanejo le skupni datumi. Eno od dobljenih zaporedij poimenuj `temp_lj_petek_13`.
    * Izpiši vse datume najbolj vročih in najbolj hladnih petkov trinajstega; pri izenačenih temperaturah ohrani vse ustrezne datume.
    * Izračunaj povprečno temperaturo ob petkih trinajstega v opazovanem obdobju.

    Namig: datumski indeks ponuja lastnosti `day_of_week` in `day`; atribut `dt` uporabljamo za datumske **vrednosti zaporedja**. Preveri, da sta poravnani zaporedji enaki in da vsak njun datum ustreza petku trinajstega.
