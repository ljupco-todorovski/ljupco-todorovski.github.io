# Knjižnica `pandas` in tabelarični podatki

[`pandas`](https://pandas.pydata.org/) je knjižnica za podatkovno analizo v Pythonu. Ponuja podatkovne strukture za shranjevanje tabelaričnih podatkov ter funkcije za njihovo pripravo, obdelavo in analizo. Podatkovne tabele so podobne preglednicam v Excelu ali tabelam v podatkovnih bazah.

Vrstice podatkovne tabele običajno ustrezajo **opazovanim enotam** (angl. _units_), za katere zbiramo podatke. Enote imenujemo tudi primerki ali primeri (angl. _examples_). Stolpci ustrezajo lastnostim opazovanih enot, ki jih imenujemo **atributi** (angl. _attributes_) ali **spremenljivke** (angl. _variables_). Vsaka celica vsebuje vrednost ene spremenljivke za eno opazovano enoto.

Za primer sestavimo podatkovno tabelo za šest oseb. Opazovane enote so osebe, izbrane spremenljivke pa ime, spol, višina v centimetrih in datum rojstva. Tabela ima šest podatkovnih vrstic; v naslovni vrstici so imena spremenljivk:

| Ime | Spol | Višina (cm) | Datum rojstva |
|:----|:----:|------------:|--------------:|
| Aleš | m | 164 | 2001-04-15 |
| Brina | ž | 153 | 2002-11-24 |
| Ciril | m | 185 | 2002-07-13 |
| Darja | ž | 181 | 2001-03-31 |
| Ema | ž | 178 | 2002-08-28 |
| Filip | m | 167 | 2002-03-09 |

Podatki v posameznem stolpcu praviloma predstavljajo vrednosti **istega tipa**. V zgornjem primeru so imena nizi, spol zavzame eno od vrednosti iz množice $\{m, ž\}$, višine so cela števila, zadnji stolpec pa vsebuje datume. Pri uvozu moramo včasih tipe določiti sami: zapis datuma v datoteki na primer še ne zagotavlja, da ga bo knjižnica prepoznala kot datum.

Knjižnica `pandas` sledi tej organizaciji podatkov. Posamezen stolpec predstavimo z **zaporedjem** (`Series`), več stolpcev pa združimo v **podatkovni okvir** (`DataFrame`), ki ga bomo imenovali tudi **podatkovna tabela**. Vsako zaporedje ima en podatkovni tip `dtype`, različni stolpci tabele pa imajo lahko različne tipe.

V nadaljevanju bomo najprej spoznali zaporedja in delo s posameznimi stolpci, nato pa podatkovne tabele. Pri ustvarjanju podatkov si bomo pomagali tudi z izpeljanimi seznami iz uvodnega dela; za računanje z obstoječimi zaporedji bomo uporabljali predvsem operacije knjižnice `pandas`.
