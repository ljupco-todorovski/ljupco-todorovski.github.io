# Zaporedja (`Series`)

[Zaporedje (`Series`)](https://pandas.pydata.org/docs/reference/api/pandas.Series.html) je enodimenzionalna podatkovna struktura knjižnice `pandas`. Podobno je seznamu v Pythonu ali stolpcu preglednice, vendar poleg vrednosti hrani tudi **indeks** in ponuja operacije za podatkovno analizo. Vsak element ima oznako v indeksu; oznake niso nujno števila in se lahko ponavljajo.

Zaporedje ima en podatkovni tip `dtype`. Ta določa način shranjevanja vrednosti; pri tipu `object` lahko zaporedje vsebuje tudi objekte različnih tipov.

Zaporedje pogosto ustvarimo iz seznama ali slovarja. Pri seznamu je privzeti indeks sestavljen iz zaporednih celih števil od $0$ do dolžine seznama minus $1$. Pri slovarju postanejo ključi oznake v indeksu, vrednosti pa elementi zaporedja.

Zaporedje ustvarimo s konstruktorjem `pd.Series`. Knjižnico `pandas` običajno uvozimo z imenom `pd`:

```python
import pandas as pd

s1 = pd.Series(range(10, 51, 10))
print(s1)
```

Spoznajmo nekaj ključnih atributov zaporedja:

  * `size`: število elementov oziroma dolžina zaporedja;
  * `dtype`: podatkovni tip zaporedja;
  * `values`: polje vrednosti elementov zaporedja;
  * `index`: oznake elementov v indeksu.

```python
print(s1.size)
print(s1.dtype)
```

Zaporedje ima pet elementov celoštevilskega tipa `int64`. Oglejmo si še atribut `values`:

```python
print(type(s1.values))
print(s1.values)
print(s1.values.tolist())
print(type(s1.values.tolist()))
```

V tem primeru je `s1.values` polje tipa `numpy.ndarray`. Njegov izpis se razlikuje od izpisa seznama: med elementi ni vejic. Z metodo `tolist()` polje pretvorimo v seznam. Seznam vrednosti lahko dobimo tudi neposredno s klicem `s1.tolist()`.

Pri drugih podatkovnih tipih lahko atribut `values` vrne tudi drugačno vrsto polja. Če izrecno potrebujemo polje knjižnice `numpy`, uporabimo metodo `to_numpy()`.

Oglejmo si še atribut `index`:

```python
print(type(s1.index))
print(s1.index)
print(s1.index.tolist())
print(type(s1.index.tolist()))
```

Indeks je v tem primeru objekt tipa `RangeIndex`. Z metodo `tolist()` njegove oznake zberemo v seznam.

Privzete oznake so cela števila od 0 do 4. Če želimo elemente poimenovati, lahko indeks nastavimo že pri ustvarjanju zaporedja ali ga pozneje zamenjamo. Število novih oznak mora biti enako dolžini zaporedja. Najprej ustvarimo kopijo `s1`, da pri spreminjanju indeksa ohranimo izvirno zaporedje:

```python
s2 = s1.copy()
s2.index = list('abcde')
# Lahko tudi s2 = pd.Series(range(10, 51, 10), index=list('abcde'))
print(s2)
```

Oznake so zdaj prve črke angleške abecede. Oglejmo si novi indeks:

```python
print(type(s2.index))
print(s2.index)
print(s2.index.tolist())
```

Zaporedje s takimi oznakami lahko ustvarimo tudi iz slovarja:

```python
s3 = pd.Series({'a': 10.0, 'b': 20.0, 'c': 30.0, 'd': 40.0, 'e': 50.0})
print(s3)
```

Zaporedji `s2` in `s3` imata enake oznake in številčno enake vrednosti, vendar različna podatkovna tipa. Zaporedje `s2` ima tip `int64`, `s3` pa `float64`. Slednji predstavlja števila s plavajočo vejico (angl. _floating point_); števil, kot je `0.1`, praviloma ne more zapisati povsem natančno.

## Indeksiranje zaporedij

Pri indeksiranju razlikujemo med **oznako** elementa in njegovim **položajem**. Izraz `s[oznaka]` izbere element po oznaki. V zaporedju `s1` je oznaka `1` celo število, v zaporedju `s2` pa istemu elementu ustreza oznaka `'b'`:

```python
print(s1[1])
print(s2["b"])
```

Za izbiranje po oznakah lahko uporabimo tudi izrecni zapis z atributom `loc`:

```python
print(s2.loc["b"])
```

Če želimo drugi element po položaju, uporabimo `iloc[1]`. V tem primeru bi lahko njegovo oznako najprej poiskali tudi v indeksu:

```python
print(s2[s2.index[1]])
print(s2.iloc[1])
```

Klic `s2.index[1]` vrne oznako drugega elementa, nato pa `s2[...]` izbere element s to oznako. Če se oznake ponavljajo, tak zapis lahko vrne več elementov. Zapis `s2.iloc[1]` vedno izbere element na položaju 1.

Za izbiro po položaju ne uporabljamo `s2[1]`: pri novejših različicah knjižnice ta zapis išče oznako `1`, ki je v `s2` ni, in sproži `KeyError`.

## Atributa `iloc` in `loc`

Atribut [`iloc`](https://pandas.pydata.org/docs/reference/api/pandas.Series.iloc.html) uporabljamo za izbiranje po **položajih**. Nenegativni položaji v zaporedju `z` so `0, 1, ..., z.size - 1`, torej vrednosti iz `range(z.size)`. Negativni položaji štejejo od konca zaporedja, tako kot pri seznamih v Pythonu:

```python
print(s2.iloc[0])
print(s2.iloc[1:3])
print(s2.iloc[1:-2])
print(s2.iloc[:])
```

Rezina `1:3` izbere elementa na položajih 1 in 2; zgornja meja 3 ni vključena. Rezina `1:-2` v našem zaporedju izbere ista elementa, `:` pa vse elemente.

Prvi rezultat je posamezna vrednost, preostali trije pa so zaporedja. Za vajo pretvori rezultate rezin v sezname.

Atribut [`loc`](https://pandas.pydata.org/docs/reference/api/pandas.Series.loc.html) uporabljamo za izbiranje po **oznakah** v indeksu. Oznake so lahko nizi, števila ali druge vrednosti; tudi celo število je pri `loc` oznaka, ne položaj:

```python
print(s2.loc['a'])
print(s2.loc['b':'d'])
print(s2.loc[:])
```

Rezina `'b':'d'` pri `loc` vključuje obe meji, torej oznake `'b'`, `'c'` in `'d'`. To se razlikuje od rezine `1:3` pri `iloc`, ki zgornje meje ne vključuje.

## Podatkovni tipi zaporedij

Vsako zaporedje ima en `dtype`, vendar to ne pomeni, da morajo vsi shranjeni objekti pripadati istemu tipu v Pythonu. To ponazorimo z združevanjem celoštevilskega zaporedja in zaporedja nizov. Funkcija `pd.concat` združi podana zaporedja in privzeto ohrani njihove oznake:

```python
s_int = pd.Series(range(3))
s_chr = pd.Series(list('abc'))
print(s_int)
print(s_chr)
```

Zaporedji združimo takole:

```python
s_cuden = pd.concat((s_int, s_chr))
print(s_cuden)
```

Združeno zaporedje ima tip `object`, ki lahko hrani tako cela števila kot nize. Cela števila se pri tem ne spremenijo v nize. Oznake v indeksu se ponavljajo, saj sta obe izvirni zaporedji imeli oznake 0, 1 in 2.

Oglejmo si izbiro po oznaki 0:

```python
print(s_cuden.loc[0])
```

Ker oznaka 0 nastopa dvakrat, rezultat vsebuje oba elementa s to oznako. Z izbiro po položaju dobimo posamezen element:

```python
print(s_cuden.iloc[3])
```

Za vajo ustvari združeno zaporedje brez ponovljenih oznak. Namig: preuči argument `ignore_index` funkcije `pd.concat`.

Pri podatkovni analizi je koristno izbrati tip, ki ustreza pomenu vrednosti. V nadaljevanju bomo spoznali kategorične in časovne podatkovne tipe.

## Kategorije (angl. _categories_)

Kategorično zaporedje vsebuje vrednosti iz določenega nabora **kategorij**, lahko pa tudi **manjkajoče vrednosti**. Manjkajoča vrednost pomeni, da za posamezen element nimamo podatka, na primer ker barva objekta ni bila zabeležena. Ne pomeni niti ničle niti dodatne kategorije. Knjižnica `pandas` manjkajoče podatke označuje na različne načine glede na podatkovni tip; v kategoričnem zaporedju jih v izpisu prikaže kot `NaN`. Njihovo prisotnost preverimo z metodo `isna()`, na primer `s_barve.isna()`, ki vrne logično zaporedje.

Primer kategorij so barve opazovanih objektov: rdeča, zelena in modra.

Barve lahko najprej shranimo kot običajno zaporedje nizov. Naključne podatke ustvarimo s funkcijami iz modula `random`:

  * `seed(42)` nastavi začetno stanje generatorja psevdonaključnih števil. Če v istem okolju nastavimo isto seme in nato izvedemo iste klice v istem vrstnem redu, dobimo iste rezultate. Število 42 je poljubno izbrano seme; funkcija sama ne izbere nobene barve. Seme običajno nastavimo enkrat pred ustvarjanjem naključnih podatkov, saj vsak nov klic `seed` ponastavi generator.
  * `choice(barve)` naključno izbere **en element** seznama. Vsak položaj v seznamu ima enako verjetnost izbire.
  * V naslednjem primeru uporabimo sorodno funkcijo `choices(barve, k=10)`, ki vrne **seznam desetih naključno izbranih elementov s ponavljanjem**. Ista barva je zato lahko izbrana večkrat. Brez podanih uteži imajo vsi položaji enako verjetnost izbire.

Ustvarimo zaporedje desetih barv:

```python
from random import choices, sample, seed

seed(42)

barve = ["rdeča", "zelena", "modra"]
s_barve = pd.Series(choices(barve, k=10))
print(s_barve)
```

Pri zaporedju nizov nabor dovoljenih vrednosti ni določen. Zato lahko barvo pomotoma zapišemo brez strešice, ne da bi dobili napako:

```python
s_barve.loc[9] = "rdeca"
print(s_barve)
```

Ustvarimo zdaj [kategorično zaporedje](https://pandas.pydata.org/docs/user_guide/categorical.html). Z objektom `CategoricalDtype` izrecno določimo vse dovoljene barve, tudi če katere pri naključnem izbiranju ne izberemo:

```python
from pandas.api.types import CategoricalDtype

kat_barve = CategoricalDtype(categories=barve)
s_barve = pd.Series(choices(barve, k=10), dtype=kat_barve)
print(s_barve)
```

Zadnja vrstica izpisa prikazuje dovoljene kategorije. Do njih dostopamo prek atributa `cat.categories`, z metodo `tolist()` pa jih pretvorimo v seznam:

```python
print(s_barve.cat.categories.tolist())
```

Prirejanje vrednosti, ki ni med dovoljenimi kategorijami, sproži izjemo `TypeError`. Izjemo v tem prikazu prestrežemo, da lahko nadaljujemo izvajanje zvezka:

```python
try:
    s_barve.loc[9] = "rdeca"
except TypeError as napaka:
    print(f"TypeError: {napaka}")
```

Prirejanje dovoljene vrednosti pa je uspešno:

```python
s_barve.loc[9] = "rdeča"
print(s_barve)
```

### Prednosti kategorij

Kategorični tip določa nabor dovoljenih vrednosti in omogoča tudi določanje njihovega vrstnega reda. Poleg tega:

  * Pri veliko ponovitvah in majhnem številu kategorij lahko zmanjša porabo pomnilnika: hrani nabor kategorij in celoštevilske kode elementov. Če je kategorij skoraj toliko kot elementov, je lahko poraba tudi večja.
  * Pri nekaterih operacijah lahko pospeši obdelavo. Hitrost je odvisna od podatkov in operacije, zato jo preverimo z meritvami.
  * Omogoča pregledno združevanje in prikaz podatkov po kategorijah, na primer štetje oseb po barvah ali računanje povprečij druge spremenljivke za posamezne kategorije.

V nadaljevanju z izrazom **kategorični tip** označujemo podatkovni tip, s **kategorijo** pa posamezno dovoljeno vrednost.

### Dodajanje in odstranjevanje kategorij

Nabor kategorij dobimo prek atributa `cat.categories`:

```python
print(s_barve.cat.categories.tolist())
```

Z metodama [`add_categories`](https://pandas.pydata.org/docs/reference/api/pandas.Series.cat.add_categories.html) in [`remove_categories`](https://pandas.pydata.org/docs/reference/api/pandas.Series.cat.remove_categories.html) dodamo ali odstranimo kategorije. Metodi vrneta novo zaporedje, zato rezultat priredimo spremenljivki:

```python
s_barve = s_barve.cat.add_categories("rumena")
print(s_barve)
s_barve = s_barve.cat.remove_categories("rdeča")
print(s_barve)
```

Za vajo pojasni, zakaj se po odstranitvi kategorije `'rdeča'` pojavijo manjkajoče vrednosti `NaN`.

### Urejene kategorije

Privzeto so kategorije **neurejene**: njihov zapisani vrstni red ne določa relacije »manjše od«. Za barve je to smiselno, pri starostnih skupinah pa lahko določimo urejenost. V našem primeru uporabimo naslednje skupine:

```python
starostne_skupine = ["otrok", "najstnica", "mladostnica", "odrasla"]
kat_ss = CategoricalDtype(categories=starostne_skupine, ordered=True)
starost = pd.Series(choices(starostne_skupine, k=10))
s_starost = starost.astype(kat_ss)
print(s_starost)
```

Znaki `<` na koncu izpisa kažejo, da so kategorije urejene v podanem vrstnem redu. Poimenovanja in meje starostnih skupin v tem primeru so izbrane za ponazoritev.

Urejene kategorije lahko ustvarimo z **diskretizacijo**: številske vrednosti razdelimo v intervale in vsakemu intervalu priredimo kategorijo. Vrednosti so lahko cela števila ali števila s plavajočo vejico. S funkcijo `pd.cut` pretvorimo starost v letih v starostne skupine:

Starosti ustvarimo s funkcijo `sample(range(100), k=10)` iz modula `random`. Ta vrne seznam desetih naključno izbranih elementov **brez ponavljanja izbranih položajev**. Ker `range(100)` vsebuje različna cela števila od 0 do 99, bodo tudi izbrane starosti različne. Število izbranih elementov `k` ne sme presegati velikosti vira. Če vir vsebuje ponovljene vrednosti, se lahko iste vrednosti pojavijo tudi v rezultatu `sample`, saj izbira različne položaje, ne nujno različnih vrednosti.

V nasprotju s `choices` torej `sample` izbira brez ponavljanja. V tem primeru uporabimo generator, katerega seme smo že nastavili ob prvem primeru z barvami: nadaljnji klici porabljajo njegovo stanje, ne začnejo vsakič znova pri istem rezultatu.

```python
starost_v_letih = pd.Series(sample(range(100), k=10))
print(starost_v_letih)
starostne_meje = [0, 12, 20, 28, float("inf")]
starost_kat = pd.cut(
    starost_v_letih, starostne_meje,
    labels=starostne_skupine, include_lowest=True
)
print(starost_kat)
```

Za vhodno zaporedje `pd.cut` vrne kategorično zaporedje, privzeto z urejenimi kategorijami. Argument `labels` poda imena kategorij, seznam mej pa mora imeti en element več kot seznam imen.

Privzeta nastavitev `right=True` pomeni, da intervali vključujejo desno mejo. Z `include_lowest=True` vključimo tudi najnižjo mejo prvega intervala. Naše skupine zato ustrezajo intervalom $[0, 12]$, $(12, 20]$, $(20, 28]$ in $(28, \infty)$. Vrednosti zunaj podanih intervalov dobijo manjkajočo vrednost.

Za vajo preveri razvrstitev starosti 0, 12, 13, 20, 21, 28 in 29. Nato preuči, kako argument `right=False` spremeni vključevanje mej.

## Datum in čas

Knjižnica `pandas` ponuja tri vrste časovnih podatkov:

  * **Časovna točka** (`Timestamp`, angl. _timestamp_) predstavlja določen trenutek, na primer datum in uro.
  * **Časovna razlika** (`Timedelta`, angl. _time delta_) predstavlja trajanje, na primer dva dneva.
  * **Časovno obdobje** (`Period`, angl. _period_) predstavlja interval z določeno frekvenco, na primer mesec oktober 2024.

### Časovne točke

S funkcijo `pd.to_datetime` pretvorimo zaporedje nizov v zaporedje časovnih točk. V primeru različnih zapisov datumov uporabimo `format="mixed"`:

```python
s_datumi = pd.to_datetime(
    pd.Series(["Jul 31, 2009", "Jan 10, 2010", "10-Oct-2023", None]),
    format="mixed"
)
print(s_datumi)
```

Nastavitev `format="mixed"` omogoča prepoznavanje zapisa za vsak element posebej in zahteva `pandas` 2.0 ali novejši. Pri dvoumnih zapisih, kot je `01/02/2024`, moramo določiti tudi, ali je prvi del dan ali mesec; kadar je mogoče, raje uporabimo enoten, izrecno določen format. Vhodna manjkajoča vrednost `None` se pretvori v `NaT`, oznako za manjkajočo časovno vrednost.

Zaporedne časovne točke ustvarimo s funkcijo `pd.date_range`, ki vrne objekt `DatetimeIndex`. V naslednjem primeru ga pretvorimo v zaporedje:

```python
s_ct1 = pd.Series(pd.date_range("2026-10-01", freq="D", periods=10))
print(s_ct1)
s_ct2 = pd.Series(pd.date_range("2026-10-01", freq="W-FRI", periods=10))
print(s_ct2)
```

Prvo zaporedje vsebuje deset zaporednih dni od 1. oktobra 2026. Drugo vsebuje deset petkov od prvega petka na izbranem območju, 2. oktobra 2026. Nastavitev `periods` določa število točk, `freq` pa njihovo frekvenco.

Časovne točke lahko uporabimo kot indeks zaporedja opazovanih vrednosti. Tako dobimo **časovno vrsto**:

```python
cv = pd.Series([i / 10 for i in range(10)], index=s_ct2)
print(cv)
```

Z datumskim indeksom lahko izberemo celotno časovno obdobje. Niz `"2026-11"` pri `loc` izbere vse vrednosti, katerih datumi so v novembru 2026:

```python
cv_novembra_26 = cv.loc["2026-11"]
print(cv_novembra_26)
```

Za zaporedje datumskih vrednosti atribut `dt` ponuja datumske lastnosti in metode. Oglejmo si dan v tednu:

```python
print(s_ct1.dt.day_of_week)
print(s_ct1.dt.day_name())
```

Lastnost `day_of_week` vedno uporablja oznake od 0 za ponedeljek do 6 za nedeljo; to ni odvisno od računalnika. Metoda `day_name()` privzeto vrne angleška imena dni.

Za vajo napiši funkcijo, ki za podano datumsko zaporedje vrne zaporedje imen dni v slovenščini. Namig: številčne oznake dni preslikaj v slovenska imena.

Prek `dt` lahko dobimo tudi številko tedna po standardu ISO in zaporedno številko dneva v letu:

```python
print(s_ct2.dt.isocalendar().week)
print(s_ct2.dt.day_of_year)
```

Definirajmo funkcijo `zadnji_dnevi`, ki za obdobje od začetka leta `od` do konca leta `do` vrne kategorično zaporedje imen dni v tednu za zadnje dni vseh mesecev. Indeks zaporedja naj vsebuje ustrezne datume. Predpostavimo celoštevilski leti `1 <= od <= do <= 9999`, katerih datumi so predstavljivi v uporabljeni različici knjižnice:

```python
def zadnji_dnevi(od, do):
    if od > do:
        raise ValueError("Začetno leto ne sme biti večje od končnega.")

    # Datumi zadnjih dni v mesecih (ME: month end).
    zdm = pd.date_range(start=f"{od:04d}-01-01", end=f"{do:04d}-12-31", freq=pd.offsets.MonthEnd())
    imena = pd.Series(zdm, index=zdm).dt.day_name()
    return imena.astype("category")


print(zadnji_dnevi(2026, 2027))
```

Za konce mesecev smo uporabili objekt `pd.offsets.MonthEnd()`. V različicah `pandas` 2.2 in novejših lahko enako frekvenco zapišemo kot `"ME"` (angl. _month end_); objekt deluje tudi v starejših različicah. Oznaka `"M"` za časovne točke je v novejših različicah opuščena.

Za vajo prilagodi funkcijo, da vrne slovenska imena dni. Določi tudi urejen kategorični tip z dnevi od ponedeljka do nedelje.

### Časovne razlike

Z odštevanjem dveh časovnih točk dobimo časovno razliko. Pri odštevanju zaporedij se elementi najprej poravnajo po oznakah; tukaj imata obe zaporedji isti privzeti indeks:


```python
s_ct2 - s_ct1
```

Rezultat je zaporedje časovnih razlik, v izpisu izraženih v dnevih. Posamezno trajanje ustvarimo s konstruktorjem [`pd.Timedelta`](https://pandas.pydata.org/docs/reference/api/pandas.Timedelta.html), pri katerem argument `unit` določa časovno enoto:

```python
s_cr1 = pd.Series([pd.Timedelta(i, unit="d") for i in range(1, 11)])
print(s_cr1)
```

S funkcijo [`pd.to_timedelta`](https://pandas.pydata.org/docs/reference/api/pandas.to_timedelta.html) pretvorimo več številskih vrednosti v časovne razlike. Te lahko uporabimo tudi kot indeks zaporedja:

```python
s_cv = pd.Series(range(10), index=pd.to_timedelta(range(1, 11), unit="s"))
print(s_cv)
```

Časovne razlike torej lahko uporabimo tako kot vrednosti kot tudi kot oznake v indeksu.

Definirajmo funkcijo `meseci`, ki za obdobje od začetka leta `od` do konca leta `do` vrne celoštevilsko zaporedje dolžin mesecev v dnevih. Indeks naj vsebuje angleška imena mesecev. Predpostavimo enake omejitve let kot pri funkciji `zadnji_dnevi`. Če obdobje obsega več let, se bodo imena mesecev v indeksu ponavljala:

```python
def meseci(od, do):
    if od > do:
        raise ValueError("Začetno leto ne sme biti večje od končnega.")

    zdm = pd.date_range(start=f"{od:04d}-01-01", end=f"{do:04d}-12-31", freq=pd.offsets.MonthEnd())
    pdm = pd.date_range(start=f"{od:04d}-01-01", end=f"{do:04d}-12-01", freq="MS")

    # Prištejemo en dan, ker upoštevamo prvi in zadnji dan meseca.
    trajanja = zdm - pdm + pd.Timedelta(days=1)
    return pd.Series(trajanja.days, index=pdm.month_name())


print(meseci(2027, 2028))
```

Lastnost `trajanja.days` pretvori časovne razlike v celoštevilsko število dni. Frekvenca `MS` označuje začetke mesecev (angl. _month start_).

Za vajo prilagodi funkcijo, da vrne dolžine mesecev v urah, nato še v tednih. Namig: časovne razlike deli s `pd.Timedelta(hours=1)` oziroma `pd.Timedelta(weeks=1)`. Pri tednih rezultat praviloma ni celo število. Indeks zamenjaj z oznakami, ki vsebujejo leto in mesec, da se pri večletnem obdobju ne ponavlja.

### Časovna obdobja

Časovna obdobja so uporabna, kadar podatek pripada celotnemu dnevu, tednu ali mesecu. V nasprotju s časovno točko obdobje predstavlja interval. S funkcijo `pd.period_range` ustvarimo zaporedna obdobja:

```python
s_co1 = pd.Series(pd.period_range("2026-10-01", freq="D", periods=10))
print(f"Dnevna obdobja:\n{s_co1}")
s_co2 = pd.Series(pd.period_range("2026-09-01", freq="W-MON", periods=10))
print(f"Tedenska obdobja s koncem v ponedeljek:\n{s_co2}")
```

Frekvenca `W-MON` pri časovnih obdobjih pomeni tedne, ki se **končajo v ponedeljek** in začnejo v torek. Ne pomeni zgolj posameznih ponedeljkov. Pri `pd.date_range` ista frekvenca izbere časovne točke ob ponedeljkih.

Za vajo primerjaj izpis dnevnih obdobij z izpisom dnevnih časovnih točk. Nato ustvari mesečna obdobja s frekvenco `M`; za mesečne časovne točke na koncu meseca pa uporabimo `ME`.
