# Naloge za vajo

Samostojne naloge utrjujejo analizo, vizualizacijo, delo z zemljevidi in razvrščanje v skupine. Krajše naloge za sprotno vajo so ob ustreznih primerih v posameznih zvezkih. Uporabi knjižnico `pandas`, uvoženo kot `pd`. Podatke in pomožne funkcije pripravi z izvajanjem primerov iz povezanih zvezkov od začetka. Naključnim generatorjem nastavi seme, pri spletnih podatkih pa preveri strukturo in referenčno obdobje vira.

## Osnovna analiza podatkov

Izhodišče: [Osnovna analiza podatkov](analiza-osnovna.ipynb).

1. **Najboljši povprečni rezultat.** Napiši program v `pandas`, ki na osnovi podatkov iz tabel `studenti` in `studenti_ocene` izpiše ime študenta ali študentke z najboljšim povprečnim rezultatom na kolokvijih. Neopravljeni kolokvij naj šteje 0 točk; ob izenačenju izpiši vsa imena.

1. **Razpon rezultatov kolokvija.** Sestavi tabelo, ki za posamezen kolokvij poroča razliko med najboljšim in najslabšim znanim rezultatom študentov.

1. **Končna tabela dosežkov.** V tabeli z rezultati kolokvijev, ki smo jo sestavili med predavanji, dodaj oziroma preveri naslednje tri spremenljivke. Neopravljeni kolokvij naj šteje 0 točk.

    * `skupaj`: skupno število doseženih točk študenta (študentke) na kolokvijih;

    * `povprecje`: povprečno število doseženih točk študenta (študentke) na kolokvijih;

    * `ocena`: dosežena ocena študenta (študentke), pri čemer upoštevate običajno ocenjevalno lestvico, kjer je 10 ocena za povprečen rezultat nad 90 točk, 9 za povprečje nad 80, 8 za povprečje nad 70, 7 za povprečje nad 60, 6 za povprečje nad 50 in 5 za povprečje največ 50.
  
1. **Vrednosti funkcije dveh spremenljivk.** Sestavi tabelo funkcije $f(x,y)=\sin(x^2+y^2)$ za vse vrednosti $x$ in $y$ na intervalu $[-2,2]$ vzorčene s korakom $0.1$, torej $\{-2, -1.9, -1.8, \ldots, 2\}$. Iz te tabele za podano vrednost $\varepsilon \ge 0$ sestavi seznam treh podatkovnih tabel: prva naj vsebuje tiste vrstice, za katere velja $z = f(x, y) < -\varepsilon$, druga vrstice $|z| \leq \varepsilon$ in tretja vrstice $z > \varepsilon$.
  
1. **Mesečni maksimum dnevnih okužb.** Spletna stran [Covid-19 sledilnik](https://covid-19.sledilnik.org/) hrani zgodovinske podatke o stanju epidemije Covid-19 v Sloveniji. Podatki o dnevnem številu okužb ponujajo tudi skozi spletne storitve [GitHub](https://github.com/sledilnik/data/blob/master/csv/stats.csv): ti podatki so v obliki CSV dostopni na naslovu [https://raw.githubusercontent.com/sledilnik/data/master/csv/stats.csv](https://raw.githubusercontent.com/sledilnik/data/master/csv/stats.csv).

    Napiši program v Pythonu s knjižnico `pandas`, ki iz teh podatkov sestavi podatkovno tabelo z urejenimi dnevnimi podatki o številu okužb v celotni Sloveniji. Podatkovna tabela naj vsebuje naslednji dve spremenljivki:
  
    * `datum` je dimenzijska spremenljivka z datumom opazovanega števila okužb;
  
    * `kumulativne_okuzbe` je merjena spremenljivka, ki za Slovenijo poda kumulativno število okužb Covid-19 do izbranega datuma.
  
    Iz dnevnih prirastov kumulativnega števila okužb sestavi novo podatkovno tabelo, ki za posamezni mesec v opazovanem obdobju izračuna maksimalno število okužb v enem dnevu tega meseca. Nova tabela naj torej vsebuje dve spremenljivki:
  
    * `mesec` je dimenzijska spremenljivka, ki določi mesec opazovanja;
  
    * `max_dnevne_okuzbe` je numerična spremenljivka, ki poda maksimalno število dnevnih okužb v izbranem mesecu.
  
    Namig: podatke najprej uredi po datumu in uporabi dnevne priraste, ne maksimuma kumulativnih vrednosti. Preveri tudi popravke podatkov, manjkajoče dneve in morebitne negativne priraste. Naloga nadaljuje nalogo o uvozu okužb iz prejšnjega poglavja.

## Vizualizacija podatkov

Izhodišče: [Vizualizacija podatkov](analiza-diagrami.ipynb).

1. **Mesečne temperature v Ljubljani.** Preberi datoteko [`temperatura.csv`](https://kt.ijs.si/~ljupco/lectures/appr-2526/temperatura.csv) s stolpci `mesec`, `max_t` in `min_t`. Zadnja dva stolpca vsebujeta mesečni povprečji dnevnih maksimalnih oziroma minimalnih temperatur v °C. Stolpec `mesec` pretvori v datum z obliko `%Y/%m`, tabelo pa z metodo `melt` iz razdelka [Tvorjenje in urejanje tabele `studenti_ocene`](urejanje-uvoz-sql.ipynb#tvorjenje-in-urejanje-tabele-studenti-ocene) uredi v stolpce `mesec`, `vrsta` in `temperatura`.

    S knjižnico `plotnine` nariši časovni potek obeh vrst temperature kot različno obarvani črti. Za vsako vrsto označi najmanjšo in največjo vrednost v celotnem obdobju, pri izenačenju pa ohrani vse ustrezne mesece. Oznake postavi z [`geom_label`](https://plotnine.org/reference/geom_label.html). Dodaj naslov, enote in razumljive oznake datumov ter diagram z metodo [`save`](https://plotnine.org/reference/ggplot.html#plotnine.ggplot.save) shrani v `temperatura.pdf`. Primerjaj z [referenčnim diagramom](https://kt.ijs.si/~ljupco/lectures/appr-2526/temperatura.png).

    Namig: ekstremne vrstice za vsako vrsto temperature izberi z združevanjem in primerjavo z minimumom oziroma maksimumom skupine. Za ohranitev vseh izenačenih ekstremov je uporabna metoda [`transform`](https://pandas.pydata.org/docs/reference/api/pandas.api.typing.SeriesGroupBy.transform.html).

1. **Plasti temperaturnega diagrama.** Preštej plasti diagrama iz prejšnje naloge, opiši podatke in geometrijske objekte vsake plasti ter označi ustrezne dele svoje kode v Pythonu. Katere estetske preslikave so globalne in katere lokalne? Pojasni, kako se tabela za plast oznak razlikuje od tabele za črti.

1. **Mesečni temperaturni razponi.** Iz iste datoteke nariši za vsak mesec navpični odsek od `min_t` do `max_t`; barva naj ponazarja razliko `max_t - min_t`. Uporabi [`geom_linerange`](https://plotnine.org/reference/geom_linerange.html) in [`scale_color_gradient`](https://plotnine.org/reference/scale_color_gradient.html). Na osi x prikaži mesec, z `facet_wrap` pa pripravi ločen vidik za vsako leto, kot v [referenčnem diagramu](https://kt.ijs.si/~ljupco/lectures/appr-2526/temperatura-razpon.png).

    Primerjaj skupne osi s `scales="free_x"`, zlasti če za zadnje leto ni podatkov za vse mesece. Os y naj omogoča primerjavo temperatur med leti. Razpon mesečnih povprečij ni isto kot razpon najnižje in najvišje dnevne meritve v mesecu.

1. **Temperature v daljšem obdobju.** V arhivu meteoroloških podatkov [ARSO](https://meteo.arso.gov.si/) poišči podatke za isto postajo za 10 ali 20 let. Zabeleži postajo, obdobje in uporabljeni spremenljivki. Če preneseš dnevne minimalne in maksimalne temperature, iz njih najprej izračunaj ustrezni mesečni povprečji; manjkajočih meritev ne nadomeščaj z ničlami.

    Ponovi črtni diagram in diagram razponov. Na črtnem diagramu tokrat označi najmanjše in največje mesečno povprečje vsake vrste **za vsako leto posebej**. Loči te ekstreme mesečnih povprečij od ekstremnih dnevnih meritev ter pojasni obravnavo nepopolnih mesecev.

## Razvrščanje v skupine

Izhodišče: [Razvrščanje v skupine](analiza-razvrscanje-v-skupine.ipynb).

1. **Trikotniška neenakost.** Za funkcijo razdalje $d$ velja trikotniška neenakost, če za vse $u$, $v$, $w$ velja $d(u,v) \leq d(u, w) + d(w, v)$. Dokaži, da trikotniška neenakost velja za evklidsko in manhattansko razdaljo. Ali velja tudi za razdaljo Minkowskega z vrednostjo $q = 1/2$?

1. **Dendrogram in skupine.** Za umetne podatke iz teh predavanj sestavi dendrogram z Wardovo mero razdalje med skupinami. Z uporabo dobljenega dendrograma razvrsti točke v štiri skupine. Sestavi podatkovno tabelo z dvema stolpcema, v prvem naj bo oznaka `id` točke, v drugem pa indeks skupine `skupina`, kjer točka pripada.

    Sestavi še eno podatkovno tabelo s štirimi stolpci, kjer prejšnji tabeli dodamo še stolpca `x` in `y` s koordinatami točke v prostoru $\mathbb{R}^2$. Uporabi to tabelo za vizualizacijo pripadnosti točk skupinam v prostoru $\mathbb{R}^2$.

1. **Telefonsko glasovanje na Evroviziji.** Članek [Eurovision Song Contest 2022](https://en.wikipedia.org/wiki/Eurovision_Song_Contest_2022) iz spletne enciklopedije Wikipedia podaja tabelo z rezultati telefonskega glasovanja držav na finalni večer pesmi Evrovizije 2022 (tabela z naslovom _Detailed televoting results of the final_).

    S pomočjo te tabele razvrstite države, ki so glasovale na finalni večer, v **štiri** skupine glede na podobnost porazdelitve točk iz telefonskega glasovanja med tekmovalnimi državami. Na zemljevidu vseh držav, ki so glasovale na finalni večer, razen Avstralije, pobarvaj posamezne države glede na njihovo pripadnost štirim skupinam.
  
    Ugotovi kateri skupini pripada Slovenija in sestavi seznam z imeni držav iz te skupine. Preveri kako se elementi seznama spreminjajo s spreminjanjem mere razdalje med skupinami v algoritmu HAC.

Pri nalogi z Evrovizijo preveri naslov in usmerjenost tabele: vrstice naj predstavljajo države, ki glasujejo, stolpci pa prejemnice točk. Odsotnost dodeljenih točk loči od neznanih podatkov. Navedi uporabljeno mero razdalje in način obravnave lastne države.
